### 1. What is Supervised Learning?
Supervised Learning is a category of Machine Learning where an algorithm learns from a **labeled dataset**. Think of it as a student learning with a teacher: the data provides the 'questions' (features) and the 'answers' (labels). The goal is for the model to learn a mapping from inputs to outputs so it can predict labels for new, unseen data.

### 2. Regression vs. Classification
Within Supervised Learning, we generally tackle two types of problems:
- **Regression:** Used when the target variable is **continuous** (numerical). Examples include predicting house prices, temperature, or stock market trends.
- **Classification:** Used when the target variable is **discrete** (categorical). Examples include identifying an email as 'Spam' or 'Not Spam', or classifying an image as a 'Cat' or 'Dog'.

### 3. The Role of Linear Regression
**Linear Regression** is the simplest and most fundamental algorithm in the Regression family. It assumes a linear relationship between the input variables ($X$) and the single output variable ($y$).

**Why start here?**
- **Interpretability:** It is easy to understand how each input affects the output.
- **Foundation:** Many advanced techniques (like Neural Networks) are built upon the principles of linear modeling.
- **Efficiency:** It is computationally inexpensive and provides a strong baseline for any predictive task.

## Module 2: Mathematical Foundations

### 1. Foundational Variables and Coordinate Systems

Before we can calculate a 'line of best fit', we must understand the space in which that line exists. In Linear Regression, we deal with two types of variables:

*   **Independent Variable ($X$):** Also known as the **Feature** or **Predictor**. This is the input we use to make a prediction. In a 2D graph, this is plotted on the **Horizontal Axis (x-axis)**.
*   **Dependent Variable ($y$):** Also known as the **Target** or **Label**. This is the outcome we are trying to predict. Its value 'depends' on $X$. In a 2D graph, this is plotted on the **Vertical Axis (y-axis)**.

#### The Cartesian Coordinate System
Every data point in our dataset can be represented as a pair $(x, y)$.
- The **x-coordinate** tells us how far right (or left) to go.
- The **y-coordinate** tells us how far up (or down) to go.

Together, these points form a 'scatter plot', which is the starting point for all linear modeling. The goal of the math we are about to learn is to find a mathematical function that passes as close to these $(x, y)$ coordinates as possible.

## Geometry of a Line

### Subtask:
Explain the concepts of Slope (steepness) and Intercept (starting point) through the equation of a straight line, including visual representations.


### 2. The Geometry of a Straight Line

In Linear Regression, our goal is to find the "line of best fit." But what mathematically defines a line? We use the **Slope-Intercept form**:

$$\Large y = mx + c$$

#### The Components:
1.  **$y$ (Dependent Variable):** The value we want to predict.
2.  **$x$ (Independent Variable):** The input value.
3.  **$m$ (Slope/Weight):** This determines the **steepness** and **direction** of the line.
    *   If $m$ is positive, the line goes up as $x$ increases.
    *   If $m$ is negative, the line goes down as $x$ increases.
    *   If $m$ is zero, the line is perfectly horizontal.
4.  **$c$ (Intercept/Bias):** This is where the line crosses the **y-axis** when $x = 0$. It represents the "starting value" or baseline.

#### Visual Intuition

Imagine we are predicting Sales ($y$) based on TV Advertising ($x$):

```text
Sales (y)
  ^
  |           / (Steep slope 'm' = high impact)
  |          /
  |         /
  |        /   
  |-------* (Intercept 'c' = Sales if TV spend is $0)
  |      /
  +------------------> TV Ads (x)
```

By adjusting $m$ and $c$, we can rotate and shift the line until it passes through our data points as accurately as possible.

## Distance Metrics

### Subtask:
Introduce Euclidean Distance and the concept of 'error' as the physical distance between a data point and a line.


### 3. Distance Metrics: Defining the Error

How do we know if a line is a "good" fit? We measure the **distance** between the actual data points and the line we've drawn. This distance represents our **Error** (also called a **Residual**).

#### Euclidean Distance
In geometry, the shortest distance between two points $(x_1, y_1)$ and $(x_2, y_2)$ is the **Euclidean Distance**, derived from the Pythagorean theorem:

$$\Large d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$

#### Error in Regression (Vertical Distance)
In Linear Regression, we are specifically interested in the **vertical distance** between the actual observed value ($y$) and the predicted value on the line ($\hat{y}$).

*   **Actual Value ($y$):** The real data point from our dataset.
*   **Predicted Value ($\hat{y}$):** The point on our line for a given $X$.
*   **Residual ($e$):** $y - \hat{y}$

#### Visualizing the Error
```text
Sales (y)
  ^
  |        * (Actual Point 'y')
  |        |
  |        | <--- Error (Vertical Distance)
  |     ___|___
  |    /   |   / (Our Line 'y-hat')
  |   /        
  +------------------> TV Ads (x)
```

A "perfect" model would have an error of zero for every point. Since that's rarely possible, our mathematical goal is to find the values of $m$ and $c$ that make the **total distance** (the sum of all errors) as small as possible.

## Functions and Linearity

### Subtask:
Define linear functions and explain why they form the simplest yet most powerful basis for regression modeling.


### 4. Functions and Linearity

At its core, Linear Regression is about finding a **Linear Function** that maps inputs to outputs. A function is simply a rule that takes an input ($X$) and produces an output ($y$).

#### What makes a Function 'Linear'?
In the context of regression, a relationship is considered linear if it follows two main principles:
1.  **Additivity:** The effect of changing one variable is independent of the values of other variables.
2.  **Proportionality (Homogeneity):** A change in the input ($X$) results in a proportional change in the output ($y$). For example, if you double your TV ad spend, a linear model assumes a consistent multiplier effect on sales.

#### Why use Linear Functions?
Linearity is a simplification of the real world, but it is incredibly powerful for several reasons:

*   **Interpretability:** Because the relationship is a straight line, we can say exactly how much $y$ will change for every unit increase in $X$. This is crucial for business decisions (e.g., 'For every extra $1,000 in ads, we expect 50 more sales').
*   **Simplicity:** Linear functions are mathematically 'well-behaved,' meaning they are easier to calculate and optimize than curvy, non-linear functions.
*   **The Foundation:** Even the most complex AI models (like Deep Learning) start with linear combinations of inputs before applying non-linear transformations. Understanding the linear base is the first step toward understanding all of Machine Learning.

#### The Assumption of Linearity
When we use Linear Regression, we are making an **assumption** that a straight line is a good representation of the data. If the data actually looks like a curve (exponential growth, for example), a simple linear model might not be enough—but it's always the best place to start!

## Module 3: Statistics Foundations

### 1. Statistical Descriptive Foundations

Before we can master regression, we must understand the basic language used to describe data. Linear Regression is built upon the concepts of 'center' and 'spread'.

#### Measures of Center
These measures help us find the 'typical' value in a dataset:
*   **Mean (Average):** The sum of all values divided by the number of values. In regression, the mean is a horizontal baseline; it represents our 'best guess' if we have no other information.
*   **Median:** The middle value when data is sorted. It is less sensitive to outliers (extremely high or low values) than the mean.

#### Measures of Spread
These measures tell us how 'stretched' or 'squeezed' our data points are:
*   **Variance:** The average of the squared differences from the Mean. It tells us how much the data fluctuates around the center.
*   **Standard Deviation:** The square root of the variance. It is expressed in the same units as the data, making it easier to interpret (e.g., if we are measuring height in cm, the standard deviation is also in cm).

#### Why these matter for Regression
Linear Regression attempts to explain the **Variance** in the dependent variable ($y$) using the independent variable ($X$). If there is no spread in our data, there is nothing for the model to predict!

## Covariance and Correlation

### Subtask:
Explain Covariance and Pearson Correlation, including formulas, ranges, and illustrative examples from the Advertising dataset.


### 2. Covariance and Correlation

Now that we can describe single variables, we need to understand how two variables **move together**. This is the heart of Linear Regression.

#### Covariance
Covariance measures the joint variability of two random variables. If the greater values of one variable mainly correspond with the greater values of the other, the covariance is positive.
*   **Formula:** $Cov(X, Y) = \frac{\sum (X_i - \bar{X})(Y_i - \bar{Y})}{n-1}$
*   **The Limitation:** Covariance tells us the *direction* (positive or negative), but its magnitude is hard to interpret because it depends on the units of the data (e.g., dollars vs. thousands of dollars).

#### Pearson Correlation Coefficient ($r$)
Correlation is a 'normalized' version of covariance. It scales the relationship to a fixed range, making it easy to compare different datasets.
*   **The Range:** Correlation always falls between **-1** and **+1**.
    *   **+1.0:** A perfect positive relationship (as $X$ goes up, $y$ goes up in a perfect line).
    *   **0.0:** No linear relationship (the variables don't seem to care about each other).
    *   **-1.0:** A perfect negative relationship (as $X$ goes up, $y$ goes down in a perfect line).

#### Practical Example: The Advertising Dataset
In our Advertising data, we can quantify these relationships:
*   **TV and Sales:** Usually show a **strong positive correlation** (e.g., $r \approx 0.9$). When TV spend increases, sales almost always increase significantly.
*   **Newspaper and Sales:** Might show a **weak positive correlation** (e.g., $r \approx 0.2$). Increasing newspaper ads has a much less predictable effect on sales.

*Correlation helps us decide which features (like TV) are the most 'useful' for our regression model!*

## Correlation vs. Causation

### Subtask:
Detail the key differences between variables moving together (correlation) and one causing another (causation).


### 3. Correlation vs. Causation

One of the most important warnings in data science is: **"Correlation does not imply causation."** Just because two variables move together doesn't mean one causes the other to happen.

#### Defining the Difference
*   **Correlation:** A statistical measure ($r$) that describes the size and direction of a relationship between two or more variables.
*   **Causation:** Indicates that one event is the result of the occurrence of the other event; i.e., there is a cause-and-effect relationship.

#### The Role of Confounding Variables
A **Confounding Variable** (or lurking variable) is an outside influence that changes the effect of a dependent and independent variable. This often creates a "spurious correlation."

**Example: Ice Cream and Sunburns**
*   Data shows that as Ice Cream sales increase, the number of Sunburns also increases.
*   There is a **strong positive correlation** between them.
*   *Does ice cream cause sunburns?* No.
*   The **Confounding Variable** is **Temperature/Sunlight**. Hot weather causes people to buy ice cream AND causes people to get sunburned.

#### Why Regression is Not Causation
Linear Regression is a powerful tool for finding patterns, but it is purely mathematical.
1.  **Directionality Problem:** A regression model might show $X$ and $y$ are related, but it cannot tell you if $X$ causes $y$, or if $y$ causes $X$.
2.  **Omitted Variables:** Your model might show a high $r$-value because both variables are being driven by a third factor you didn't include in your dataset.

#### Implications for the California Housing Dataset
In the California Housing data, we might find a correlation between `Median Income` and `House Value`. While it's likely that higher income allows people to buy more expensive houses, we must also consider factors like local school quality or proximity to the ocean, which might be the true drivers of price and attract high-income residents simultaneously.

### 4. Visualizing Relationships

Numbers like the correlation coefficient ($r$) are powerful, but visualization is essential to truly understand the data. Let's look at how different relationships appear in a scatter plot.

#### Correlation Types

**A. Strong Positive Correlation ($r \approx +0.9$)**
As $X$ increases, $y$ increases consistently.
```text
y |
  |           *
  |        * *
  |      * *
  |    * *

```

**B. Strong Negative Correlation ($r \approx -0.9$)**
As $X$ increases, $y$ decreases consistently.
```text

  |  * *
  |    * *
  |      * *
  |        * *
  |           *
  +----------------- x
```
*Regression Insight:* A line with a negative slope will fit this data very well.

**C. Zero / No Correlation ($r \approx 0.0$)**
There is no apparent relationship between $X$ and $y$.
```text
y |

  |  *    *      *
```

*Regression Insight:* Standard Linear Regression will fail to capture this pattern. We would need to transform the data (like using polynomial features) to model this curve.

#### 1. Manual Calculation Steps

$$\bar{y} = \frac{5+9+7}{3} = 7$$

**B. Calculate Variance of $X$ ($s^2_x$):**

# Module 4: The Regression Equation

### 1. Definition & Notation
In formal statistics, we move away from $y = mx + c$ to a notation that allows us to scale to many features easily.

**The Population Equation:**
$$y = \beta_0 + \beta_1x + \epsilon$$

**The Prediction Equation (The Model):**
$$\hat{y} = \beta_0 + \beta_1x$$

#### Symbol Breakdown:
*   **$\hat{y}$ (y-hat):** The **Predicted value**. This is the output our model produces.
*   **$y$:** The **Observed value**. The actual truth from our dataset.
*   **$\beta_0$ (Beta-zero):** The **Intercept**. The value of $y$ when $x=0$.
*   **$\beta_1$ (Beta-one):** The **Coefficient** (Slope). The amount $y$ changes for a 1-unit change in $x$.
*   **$\epsilon$ (Epsilon):** The **Error term**. This represents the 'noise' or factors we didn't measure.

### 2. Mathematics of Residuals
A **Residual ($e$)** is the difference between what actually happened and what our model predicted.

**Formula:**
$$e_i = y_i - \hat{y}_i$$

#### Intuition:
*   If $e > 0$: The model **under-estimated** (the actual value is above the line).
*   If $e < 0$: The model **over-estimated** (the actual value is below the line).
*   If $e = 0$: The prediction was **perfect**.

### 3. Visual Explanation (ASCII)
```text
Sales
  ^      * (Actual y)
  |      |
  |      | <--- Residual (e = y - y_hat)
  |______|_____
  |      . (Predicted y_hat on the line)
  |     /
  +------------------> TV Budget
```

# Module 5: How Linear Regression Learns (Optimization)

### 1. The Goal: Minimizing Loss
Now that we know what residuals are, we need a way to find the **best** $\beta_0$ and $\beta_1$. In machine learning, we do this by defining a **Loss Function** (also called a Cost Function).

### 2. Mean Squared Error (MSE)
The most common loss function for regression is **Mean Squared Error**. We square the residuals so that positive and negative errors don't cancel each other out, and to penalize larger errors more heavily.

$$MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

### 3. Gradient Descent: The Intuition
Imagine you are standing on a foggy mountain (the 'Error Hill') and want to find the valley (the minimum error).
1. You feel the slope under your feet (**The Gradient**).
2. You take a small step in the downward direction (**The Learning Rate**).
3. You repeat this until you reach the bottom.

## Mathematical Derivation of the Gradient

### Subtask:
Explain the calculus behind finding the direction to minimize the cost function.


### 3. Mathematical Derivation of the Gradient

To move 'downhill' toward the minimum error, we need to know the slope of the cost function at our current position. In calculus, this is done using **Partial Derivatives**.

#### The Cost Function (MSE)
Recall our Mean Squared Error formula:
$$J(\beta_0, \beta_1) = \frac{1}{n} \sum_{i=1}^{n} (y_i - (\beta_0 + \beta_1 x_i))^2$$

#### Step 1: Derivative with respect to $\beta_0$ (The Intercept)
We apply the power rule and the chain rule to the MSE formula:
$$\frac{\partial J}{\partial \beta_0} = \frac{2}{n} \sum_{i=1}^{n} (y_i - (\beta_0 + \beta_1 x_i)) \cdot (-1)$$
Simplifying, we get the gradient for the intercept:
$$\frac{\partial J}{\partial \beta_0} = -\frac{2}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)$$

#### Step 2: Derivative with respect to $\beta_1$ (The Slope)
Following the same rules, but considering the $x_i$ attached to $\beta_1$:
$$\frac{\partial J}{\partial \beta_1} = \frac{2}{n} \sum_{i=1}^{n} (y_i - (\beta_0 + \beta_1 x_i)) \cdot (-x_i)$$
Simplifying, we get the gradient for the slope:
$$\frac{\partial J}{\partial \beta_1} = -\frac{2}{n} \sum_{i=1}^{n} x_i(y_i - \hat{y}_i)$$

#### What does the Gradient tell us?
- The **Gradient Vector** $\nabla J = [\frac{\partial J}{\partial \beta_0}, \frac{\partial J}{\partial \beta_1}]$ points in the direction of the *steepest increase* in error.
- To minimize error, we move in the **opposite direction** of the gradient.
- If the gradient is negative, we increase the parameter; if it is positive, we decrease it.

## Optimization via Gradient Descent Implementation

### Subtask:
Create a full Python loop that implements Gradient Descent from scratch and track the loss curve.


**Reasoning**:
I will implement the Gradient Descent algorithm in Python using the sample advertising data and track the MSE over 100 iterations.



````python
import numpy as np
import matplotlib.pyplot as plt

# Data (TV Budget and Sales)
X = np.array([100, 150, 200, 250, 300])
y = np.array([15, 12, 19, 22, 18])
n = len(X)

# 1. Initialize Parameters
beta_0 = 0.0
beta_1 = 0.0
learning_rate = 0.00001 # Small alpha to prevent divergence due to X scale
epochs = 100

loss_history = []

# 2. Gradient Descent Loop
for i in range(epochs):
    # Predictions
    y_hat = beta_0 + beta_1 * X

    # Calculate MSE (Cost)
    mse = np.mean((y - y_hat)**2)
    loss_history.append(mse)

    # Calculate Gradients
    # dJ/db0 = -2/n * sum(y - y_hat)
    # dJ/db1 = -2/n * sum(x * (y - y_hat))
    d_beta_0 = (-2/n) * np.sum(y - y_hat)
    d_beta_1 = (-2/n) * np.sum(X * (y - y_hat))

    # Update Parameters
    beta_0 = beta_0 - (learning_rate * d_beta_0)
    beta_1 = beta_1 - (learning_rate * d_beta_1)

# 3. Plot the Loss Curve
plt.figure(figsize=(10, 5))
plt.plot(range(epochs), loss_history, color='blue', linewidth=2)
plt.title('Loss Curve (Convergence of Gradient Descent)')
plt.xlabel('Iteration (Epoch)')
plt.ylabel('Mean Squared Error (MSE)')
plt.grid(True)
plt.show()

print(f"Final Parameters after {epochs} epochs:")
print(f"Intercept (Beta 0): {beta_0:.4f}")
print(f"Slope (Beta 1): {beta_1:.4f}")
print(f"Final MSE: {loss_history[-1]:.4f}")
````

## Advanced Optimization: Learning Rates and Convergence

### Subtask:
Demonstrate how different learning rates (alpha) affect the speed and stability of model convergence using the California Housing dataset.


**Reasoning**:
I will load the California Housing dataset, isolate a single feature for simplicity, and define a reusable Gradient Descent function to test multiple learning rates.



````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Load the California Housing dataset
df_housing = pd.read_csv('/content/sample_data/california_housing_train.csv')

# Use 'median_income' to predict 'median_house_value'
# Scale house value to thousands for better numerical stability during GD
X = df_housing['median_income'].values
y = df_housing['median_house_value'].values / 1000.0
n = len(X)

def run_gradient_descent(X, y, alpha, epochs=50):
    b0, b1 = 0.0, 0.0
    history = []
    for _ in range(epochs):
        y_hat = b0 + b1 * X
        mse = np.mean((y - y_hat)**2)
        history.append(mse)

        # Stop if MSE becomes astronomical (divergence)
        if mse > 1e10: break

        db0 = (-2/n) * np.sum(y - y_hat)
        db1 = (-2/n) * np.sum(X * (y - y_hat))
        b0 -= alpha * db0
        b1 -= alpha * db1
    return history

# 2. Test different learning rates
lrs = [0.0001, 0.01, 0.08]
labels = ['Too Small (0.0001)', 'Optimal (0.01)', 'Too Large (0.08)']

plt.figure(figsize=(12, 6))
for lr, label in zip(lrs, labels):
    loss_hist = run_gradient_descent(X, y, lr)
    plt.plot(loss_hist, label=f'Alpha: {label}')

plt.yscale('log') # Log scale to see detail when divergence happens
plt.title('Effect of Learning Rate on Convergence (California Housing Data)')
plt.xlabel('Epochs')
plt.ylabel('MSE (Log Scale)')
plt.legend()
plt.grid(True, which='both', linestyle='--', alpha=0.5)
plt.show()

print('As seen in the plot: a small alpha converges slowly, an optimal alpha reaches the minimum quickly, and a large alpha causes the error to explode (overshooting).')
````

## 3D Residual Analysis

### Subtask:
Visualize the vertical distances (residuals) from actual data points to a fitted regression plane in 3D space.


**Reasoning**:
I will calculate the coefficients for the regression plane using the Normal Equation, generate a grid to represent the plane, and then plot the scatter points, the plane, and the residual lines (stems).



````python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# 1. Prepare data for the Normal Equation (Xw = y)
# We need a column of 1s for the intercept
X_mat = np.column_stack((np.ones(len(tv)), tv, radio))
y_mat = sales.reshape(-1, 1)

# 2. Solve for weights (w) using the Normal Equation: w = (X^T X)^-1 X^T y
weights = np.linalg.inv(X_mat.T @ X_mat) @ X_mat.T @ y_mat
b0_hat, b1_hat, b2_hat = weights.flatten()

# 3. Create a grid for the plane
tv_range = np.linspace(0, 300, 20)
radio_range = np.linspace(0, 50, 20)
TV_grid, RADIO_grid = np.meshgrid(tv_range, radio_range)
SALES_plane = b0_hat + b1_hat * TV_grid + b2_hat * RADIO_grid

# 4. Calculate predictions for actual points to find residuals
sales_pred = b0_hat + b1_hat * tv + b2_hat * radio

# 5. Plotting
fig = plt.figure(figsize=(12, 9))
ax = fig.add_subplot(111, projection='3d')

# Plot the plane
ax.plot_surface(TV_grid, RADIO_grid, SALES_plane, alpha=0.3, color='cyan', label='Regression Plane')

# Plot the scatter points
ax.scatter(tv, radio, sales, c='blue', s=30, label='Actual Sales')

# Plot the residuals (vertical lines)
for i in range(len(tv)):
    ax.plot([tv[i], tv[i]], [radio[i], radio[i]], [sales[i], sales_pred[i]], color='red', alpha=0.5, linewidth=1)

ax.set_xlabel('TV Advertising Budget ($)')
ax.set_ylabel('Radio Advertising Budget ($)')
ax.set_zlabel('Sales (Units)')
ax.set_title('3D Residual Analysis: Visualizing Error as Vertical Distance to a Plane')

print(f"Fitted Equation: Sales = {b0_hat:.2f} + {b1_hat:.3f}*TV + {b2_hat:.3f}*Radio")
plt.show()
````

## The Least Squares Method (Normal Equation)

### Subtask:
Introduce and implement the Normal Equation as an analytical alternative to Gradient Descent for solving Linear Regression.


### 1. Analytical vs. Iterative Solutions

Until now, we have used **Gradient Descent**, an iterative optimization algorithm that takes small steps to reach the bottom of the error bowl. However, for Linear Regression, there is a direct way to calculate the optimal parameters in a single step using linear algebra. This is known as the **Normal Equation**.

#### The Conceptual Difference:
*   **Gradient Descent:** Like walking down a mountain in the fog; you look at the local slope and move downward repeatedly.
*   **Normal Equation:** Like having a map of the entire mountain and a formula that points directly to the exact coordinates of the valley floor.

### 2. The Normal Equation Formula

To solve for the vector of weights $\theta$ (which includes our intercept $\beta_0$ and slopes $\beta_1, \beta_2...$), we use the following matrix equation:

$$\Large \theta = (X^T X)^{-1} X^T y$$

#### Component Breakdown:
*   **$X$:** The feature matrix (with a column of 1s added for the intercept).
*   **$X^T$:** The transpose of the feature matrix.
*   **$(X^T X)^{-1}$:** The inverse of the product of the transpose and the original matrix.
*   **$y$:** The vector of actual target values.

### 3. Why use the Normal Equation?
*   **No Hyperparameters:** You don't need to choose a learning rate ($\alpha$) or number of iterations.
*   **Exact Solution:** It provides the mathematically precise global minimum of the MSE cost function.
*   **Efficiency for Small Data:** For datasets with a few thousand features, it is often faster than Gradient Descent. However, because calculating a matrix inverse is computationally expensive ($O(n^3)$), Gradient Descent is preferred for very large datasets (millions of features).

**Reasoning**:
I will implement the Normal Equation formula using NumPy's linear algebra functions and compare the resulting coefficients with those obtained from the previous iterative Gradient Descent run.



````python
import numpy as np

# 1. Prepare the data for matrix operations
# We use the 'tv' and 'sales' variables from previous steps
# X needs a column of ones for the intercept (Beta 0)
X = np.column_stack((np.ones(len(tv)), tv))
y = sales.reshape(-1, 1)

# 2. Implement the Normal Equation: theta = (X^T * X)^-1 * X^T * y
# np.linalg.inv calculates the matrix inverse
# @ is the symbol for matrix multiplication in Python
theta = np.linalg.inv(X.T @ X) @ X.T @ y

b0_normal = theta[0][0]
b1_normal = theta[1][0]

print("--- Normal Equation Results ---")
print(f"Optimal Intercept (Beta 0): {b0_normal:.4f}")
print(f"Optimal Slope (Beta 1):     {b1_normal:.4f}")

# 3. Compare with our previous Gradient Descent run
print(f"\n--- Gradient Descent Comparison (After 50 epochs) ---")
print(f"GD Intercept: {beta_0:.4f}")
print(f"GD Slope:     {beta_1:.4f}")

print(f"\nNote: The Normal Equation provides the exact mathematical minimum instantly, whereas GD requires many iterations to converge to these same values.")
````

## Mathematical Intuition of the Gradient

### Subtask:
Define the Gradient vector and explain the calculus-based reasoning for parameter minimization using the negative gradient.


### 1. Mathematical Intuition of the Gradient

To optimize a Linear Regression model, we need an automated way to find the values of $\beta_0$ and $\beta_1$ that result in the lowest possible error. This is where the **Gradient** comes in.

#### The Gradient Vector ($\nabla J$)
In multi-dimensional space (like our MSE 'bowl'), the **Gradient** is a vector that contains all the partial derivatives of the cost function $J$ with respect to each parameter:

$$\nabla J(\beta_0, \beta_1) = \begin{bmatrix} \frac{\partial J}{\partial \beta_0} \\ \frac{\partial J}{\partial \beta_1} \end{bmatrix}$$

#### Why the Gradient Matters
1.  **Direction of Steepest Ascent:** Mathematically, the gradient vector always points in the direction where the function increases most rapidly. If you were standing on the side of a mountain, the gradient would point directly 'up' the steepest part of the slope.
2.  **The Minimization Rule:** Since our goal is to **minimize** the cost, we want to go in the exact opposite direction. This is why we move in the direction of the **negative gradient**.

#### The Update Rule
To update our parameters, we take the current value and subtract a portion of the gradient:

$$\beta_{new} = \beta_{old} - \alpha \cdot \nabla J$$

*   **$\alpha$ (Learning Rate):** A small multiplier that determines how big of a step we take.
*   **The Minus Sign:** This is the critical part that ensures we are moving 'downhill' toward the minimum.

#### Geometric Analogy: The Foggy Mountain
Imagine you are lost on a mountain in a thick fog. You can't see the valley at the bottom, but you can feel the slope of the ground under your feet.
- The **Gradient** tells you which way is 'up'.
- To find the valley, you take a step in the **opposite direction** of the steepness you feel.
- As the ground becomes flatter (the gradient approaches zero), you know you are reaching the bottom (the optimal parameters).

## Step-by-Step Derivation of Partial Derivatives

### Subtask:
Perform the formal mathematical derivation of the MSE cost function partial derivatives with respect to Beta 0 and Beta 1.


### 2. Step-by-Step Derivation of Partial Derivatives

To implement Gradient Descent, we need to calculate exactly how a change in our parameters ($eta_0, eta_1$) affects our total error. This is done by taking the **partial derivative** of the Cost Function.

#### The Starting Point: MSE Cost Function
$$J(eta_0, eta_1) = \frac{1}{n} \sum_{i=1}^{n} (y_i - (\beta_0 + eta_1 x_i))^2$$

---

#### Part A: Deriving the Gradient for $\beta_0$ (Intercept)
We want to find $\frac{\partial J}{\partial \beta_0}$.

1. **Power Rule:** Bring the exponent (2) to the front.
   $$\frac{\partial J}{\partial \beta_0} = \frac{2}{n} \sum_{i=1}^{n} (y_i - (\beta_0 + \beta_1 x_i))^1 \cdot \frac{\partial}{\partial \beta_0}(y_i - (\beta_0 + \beta_1 x_i))$$

2. **Chain Rule:** The derivative of the inside with respect to $\beta_0$ is **-1** (since $y_i, \beta_1,$ and $x_i$ are treated as constants).
   $$\frac{\partial J}{\partial \beta_0} = \frac{2}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i) \cdot (-1)$$

3. **Final Form:**
   $$\frac{\partial J}{\partial \beta_0} = -\frac{2}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)$$

---

#### Part B: Deriving the Gradient for $\beta_1$ (Slope)
We want to find $\frac{\partial J}{\partial \beta_1}$.

1. **Power Rule:**
   $$\frac{\partial J}{\partial \beta_1} = \frac{2}{n} \sum_{i=1}^{n} (y_i - (\beta_0 + \beta_1 x_i)) \cdot \frac{\partial}{\partial \beta_1}(y_i - (\beta_0 + \beta_1 x_i))$$

2. **Chain Rule:** The derivative of the inside with respect to $\beta_1$ is **$-x_i$** (because $\beta_1$ is multiplied by $x_i$).
   $$\frac{\partial J}{\partial \beta_1} = \frac{2}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i) \cdot (-x_i)$$

3. **Final Form:**
   $$\frac{\partial J}{\partial \beta_1} = -\frac{2}{n} \sum_{i=1}^{n} x_i(y_i - \hat{y}_i)$$

---

#### The Update Equations
Using these gradients, we arrive at the algorithmic steps for each iteration of training:

**Update Intercept:** $\beta_0 = \beta_0 - \alpha \cdot \left[ -\frac{2}{n} \sum (y_i - \hat{y}_i) \right]$

**Update Slope:** $\beta_1 = \beta_1 - \alpha \cdot \left[ -\frac{2}{n} \sum x_i(y_i - \hat{y}_i) \right]$

## Gradient Descent Mechanics & Hyperparameters

### Subtask:
Define and explain the roles of Epochs, Iterations, and the Learning Rate (Alpha) in the optimization process.


### 3. Gradient Descent Mechanics & Hyperparameters

To move from the math of derivatives to a working computer program, we need to define the 'settings' or **hyperparameters** that control how our algorithm runs.

#### 1. The Learning Rate (Alpha - $\alpha$)
This is the most critical hyperparameter in machine learning. It determines the **size of the steps** we take down the error surface.
- **Small $\alpha$:** The steps are tiny and precise. The model is stable and likely to find the minimum, but it will take a very long time (high computational cost).
- **Large $\alpha$:** The steps are huge. The model might reach the minimum faster, but it risks 'overshooting' and bouncing out of the valley, causing the error to increase instead of decrease (divergence).

#### 2. Epochs
An **Epoch** is one complete pass through the entire training dataset.
- In each epoch, the algorithm looks at every data point, calculates the total error, and updates the weights ($eta_0, eta_1$) once.
- Training usually requires many epochs (e.g., 100, 1,000, or more) for the line to gradually rotate and settle into the 'best fit' position.

#### 3. Iterations
While an epoch is a full pass of the data, an **Iteration** is a single update of the model's weights.
- In **Batch Gradient Descent** (which we are focusing on), one Epoch equals one Iteration because we use all the data to make one single update.
- In more advanced versions (like Stochastic or Mini-batch GD), an epoch is split into multiple iterations as the weights are updated more frequently using smaller slices of the data.

#### Visualizing the Trade-off
Think of the optimization as a game of golf:
- The **Gradient** tells you the direction of the hole.
- The **Learning Rate** is the force of your swing.
- An **Epoch** is a single stroke.
- If your swing is too hard (large $\alpha$), you'll fly right over the hole. If it's too soft (small $\alpha$), you'll have to take 100 putts to get there.

## Robust Implementation with Stopping Criteria

### Subtask:
Implement a Python Gradient Descent function that monitors the change in loss between iterations and incorporates 'Tolerance' as a stopping criterion.


**Reasoning**:
I will implement the robust Gradient Descent function with a tolerance check to optimize the parameters and prevent unnecessary iterations after convergence.



````python
import numpy as np

def gradient_descent_robust(X, y, alpha=0.00001, epochs=1000, tolerance=1e-6):
    # Initialize parameters
    b0, b1 = 0.0, 0.0
    n = len(X)
    loss_history = []

    for i in range(epochs):
        # Predictions
        y_hat = b0 + b1 * X

        # Current Loss (MSE)
        mse = np.mean((y - y_hat)**2)
        loss_history.append(mse)

        # Early Stopping check
        if i > 0:
            if abs(loss_history[-2] - loss_history[-1]) < tolerance:
                print(f'Convergence reached at epoch {i} with loss change < {tolerance}')
                break

        # Gradients
        db0 = (-2/n) * np.sum(y - y_hat)
        db1 = (-2/n) * np.sum(X * (y - y_hat))

        # Update parameters
        b0 -= alpha * db0
        b1 -= alpha * db1

    return b0, b1, loss_history

# Using the 'tv' and 'sales' variables from the previous notebook state
final_b0, final_b1, history = gradient_descent_robust(tv, sales, alpha=0.00002, epochs=2000, tolerance=1e-5)

print(f'Final Intercept (Beta 0): {final_b0:.4f}')
print(f'Final Slope (Beta 1): {final_b1:.4f}')
print(f'Final MSE: {history[-1]:.4f}')
````

### Advanced Visualization: Batch vs. Stochastic Trajectories

In our previous plots, we saw a smooth descent. However, in modern deep learning, we often use **Stochastic Gradient Descent (SGD)**.

Instead of calculating the gradient using the *entire* dataset (Batch), SGD calculates the gradient for a **single random point**. This makes the optimization path look like a 'noisy' or 'drunken' walk. While it looks less efficient, this noise actually helps models escape 'local minima' in more complex problems.

````python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# 1. Setup Parameters
np.random.seed(42)
n_points = len(tv_scaled)
b0, b1 = -0.8, 0.8 # Start at a high-error corner
alpha = 0.1

path_batch = [[b0, b1, np.mean((sales_scaled - (b0 + b1 * tv_scaled))**2)]]
path_sgd = [[b0, b1, np.mean((sales_scaled - (b0 + b1 * tv_scaled))**2)]]

# 2. Simulate Batch GD (Smooth)
b0_b, b1_b = b0, b1
for _ in range(20):
    y_hat = b0_b + b1_b * tv_scaled
    db0 = (-2/n_points) * np.sum(sales_scaled - y_hat)
    db1 = (-2/n_points) * np.sum(tv_scaled * (sales_scaled - y_hat))
    b0_b -= alpha * db0
    b1_b -= alpha * db1
    path_batch.append([b0_b, b1_b, np.mean((sales_scaled - (b0_b + b1_b * tv_scaled))**2)])

# 3. Simulate Stochastic GD (Noisy)
b0_s, b1_s = b0, b1
for _ in range(50): # More steps because they are smaller/noisier
    idx = np.random.randint(0, n_points)
    x_i, y_i = tv_scaled[idx], sales_scaled[idx]
    y_hat_i = b0_s + b1_s * x_i
    # Gradient based on just ONE point
    db0_s = -2 * (y_i - y_hat_i)
    db1_s = -2 * x_i * (y_i - y_hat_i)
    b0_s -= (alpha * 0.5) * db0_s # Lower alpha for SGD stability
    path_sgd.append([b0_s, b1_s, np.mean((sales_scaled - (b0_s + b1_s * tv_scaled))**2)])

# 4. Visualization
path_batch = np.array(path_batch)
path_sgd = np.array(path_sgd)

fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

# Reuse the scaled MSE surface from kernel state B0_s, B1_s, Z_s
ax.plot_surface(B0_s, B1_s, Z_s, cmap='viridis', alpha=0.3)

# Plot paths
ax.plot(path_batch[:,0], path_batch[:,1], path_batch[:,2], color='blue', linewidth=3, label='Batch GD (Smooth)', zorder=30)
ax.plot(path_sgd[:,0], path_sgd[:,1], path_sgd[:,2], color='red', linestyle='--', alpha=0.8, label='Stochastic GD (Noisy)', zorder=25)

ax.set_title("Batch vs. Stochastic Gradient Descent Paths")
ax.set_xlabel("Beta 0")
ax.set_ylabel("Beta 1")
ax.legend()
ax.view_init(elev=20, azim=45)
plt.show()
````

### Quantitative Comparison: Normal Equation vs. Gradient Descent

We will now measure the actual execution time for both methods across different feature counts. Note how the Normal Equation eventually becomes significantly slower due to the $O(n^3)$ cost of matrix inversion.

````python
import time
import numpy as np
import matplotlib.pyplot as plt

def benchmark_methods():
    feature_counts = [100, 500, 1000, 2000, 3500, 5000]
    normal_times = []
    gd_times = []

    for n in feature_counts:
        # Generate synthetic data with n features
        X = np.random.randn(10000, n)
        y = np.random.randn(10000, 1)

        # 1. Benchmark Normal Equation
        start = time.time()
        X_bias = np.column_stack((np.ones(len(X)), X))
        theta = np.linalg.inv(X_bias.T @ X_bias) @ X_bias.T @ y
        normal_times.append(time.time() - start)

        # 2. Benchmark Gradient Descent (10 iterations for comparison scale)
        # We only run a few iterations to represent the cost per update
        start = time.time()
        beta = np.zeros((n + 1, 1))
        for _ in range(10):
            y_hat = X_bias @ beta
            gradient = (2/len(X)) * X_bias.T @ (y_hat - y)
            beta -= 0.01 * gradient
        gd_times.append(time.time() - start)

    # Plotting
    plt.figure(figsize=(10, 6))
    plt.plot(feature_counts, normal_times, marker='o', label='Normal Equation O(n³)', linewidth=2)
    plt.plot(feature_counts, gd_times, marker='s', label='Gradient Descent (10 Iterations)', linewidth=2)
    plt.xlabel('Number of Features (n)')
    plt.ylabel('Execution Time (seconds)')
    plt.title('Scalability: Normal Equation vs. Gradient Descent')
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.show()

benchmark_methods()
````

### Visualizing Learning Rate Scenarios

In this experiment, we use the **California Housing** dataset. We will compare:
1.  **Tiny Alpha:** Extremely slow convergence.
2.  **Optimal Alpha:** Fast and stable descent.
3.  **Aggressive Alpha:** Overshooting the minimum, leading to mathematical 'explosion' (NaN values).

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Load and Standardize Data
df = pd.read_csv('/content/sample_data/california_housing_train.csv')
X = df['median_income'].values
y = (df['median_house_value'].values / 1000.0) # Scale y to thousands

# Standardize X for stability
X_scaled = (X - np.mean(X)) / np.std(X)
n = len(X_scaled)

def gd_experiment(alpha, epochs=100):
    b0, b1 = 0.0, 0.0
    losses = []
    for i in range(epochs):
        y_hat = b0 + b1 * X_scaled
        mse = np.mean((y - y_hat)**2)
        losses.append(mse)

        # Stop if exploding
        if np.isinf(mse) or mse > 1e10:
            losses = losses + [losses[-1]] * (epochs - len(losses))
            break

        db0 = (-2/n) * np.sum(y - y_hat)
        db1 = (-2/n) * np.sum(X_scaled * (y - y_hat))
        b0 -= alpha * db0
        b1 -= alpha * db1
    return losses

# 2. Run Simulations
results = {
    'Tiny (0.001)': gd_experiment(0.001),
    'Optimal (0.1)': gd_experiment(0.1),
    'Explosive (1.1)': gd_experiment(1.1)
}

# 3. Plotting the results
plt.figure(figsize=(12, 6))
for label, loss_path in results.items():
    plt.plot(loss_path, label=label, linewidth=2)

plt.yscale('log')
plt.title('Impact of Learning Rate on Convergence Stability')
plt.xlabel('Epochs')
plt.ylabel('MSE (Log Scale)')
plt.legend()
plt.grid(True, which='both', linestyle='--', alpha=0.5)
plt.show()
````

## Why Scaling is Mandatory: The Geometry of Convergence

When features have vastly different scales (e.g., income vs. number of rooms), the MSE cost function becomes an **elongated, narrow valley**.

- **In the Unscaled Surface:** The gradient is very steep in one direction and very flat in another. This causes Gradient Descent to 'bounce' or 'zigzag' off the walls of the valley.
- **In the Scaled Surface:** Standardization transforms the landscape into a symmetric **symmetric bowl**, allowing the algorithm to step directly toward the center with high stability.

````python
def get_mse_surface(x, y, b0_r, b1_r):
    B0, B1 = np.meshgrid(b0_r, b1_r)
    MSE = np.array([np.mean((y - (b0 + b1 * x))**2) for b0, b1 in zip(np.ravel(B0), np.ravel(B1))])
    return B0, B1, MSE.reshape(B0.shape)

fig = plt.figure(figsize=(16, 7))

# 1. Unscaled Surface (using raw Median Income)
ax1 = fig.add_subplot(121, projection='3d')
# Center around a reasonable estimate for raw data
B0_u, B1_u, Z_u = get_mse_surface(X, y, np.linspace(-50, 50, 40), np.linspace(30, 50, 40))
ax1.plot_surface(B0_u, B1_u, Z_u, cmap='magma', alpha=0.8)
ax1.set_title("Unscaled: Elongated 'Zigzag' Valley")
ax1.set_xlabel('Beta 0')
ax1.set_ylabel('Beta 1')

# 2. Scaled Surface (using Standardized Income)
ax2 = fig.add_subplot(122, projection='3d')
B0_s, B1_s, Z_s = get_mse_surface(X_scaled, y, np.linspace(150, 250, 40), np.linspace(70, 90, 40))
ax2.plot_surface(B0_s, B1_s, Z_s, cmap='viridis', alpha=0.8)
ax2.set_title("Scaled: Symmetric 'Stable' Bowl")
ax2.set_xlabel('Beta 0')
ax2.set_ylabel('Beta 1')

plt.tight_layout()
plt.show()
````

## Regularization: Ridge (L2) and Lasso (L1)

Regularization adds a penalty term to the MSE cost function to prevent coefficients from becoming too large, which reduces overfitting.

- **Ridge Regression (L2):** Adds $\lambda \sum \beta^2$. It shrinks coefficients toward zero but rarely makes them exactly zero.
- **Lasso Regression (L1):** Adds $\lambda \sum |\beta|$. It can force coefficients to be exactly zero, performing automatic feature selection.

````python
def gradient_descent_reg(X, y, alpha=0.01, lmbda=0.1, epochs=500, mode='ridge'):
    b0, b1 = 0.0, 0.0
    n = len(X)
    for _ in range(epochs):
        y_hat = b0 + b1 * X
        db0 = (-2/n) * np.sum(y - y_hat)

        if mode == 'ridge':
            # Gradient of MSE + lambda * beta^2 is 2*lambda*beta
            db1 = (-2/n) * np.sum(X * (y - y_hat)) + 2 * lmbda * b1
        elif mode == 'lasso':
            # Gradient of MSE + lambda * |beta| is lambda * sign(beta)
            db1 = (-2/n) * np.sum(X * (y - y_hat)) + lmbda * np.sign(b1)
        else:
            db1 = (-2/n) * np.sum(X * (y - y_hat))

        b0 -= alpha * db0
        b1 -= alpha * db1
    return b0, b1

# Compare weights with a more stable alpha and lower lambda
b_ols = gradient_descent_reg(X_scaled, y, mode='ols', alpha=0.01)
b_ridge = gradient_descent_reg(X_scaled, y, mode='ridge', lmbda=0.1, alpha=0.01)
b_lasso = gradient_descent_reg(X_scaled, y, mode='lasso', lmbda=0.1, alpha=0.01)

print(f'OLS Slope:   {b_ols[1]:.4f}')
print(f'Ridge Slope: {b_ridge[1]:.4f} (Shrunk by L2 penalty)')
print(f'Lasso Slope: {b_lasso[1]:.4f} (Shrunk by L1 penalty)')
````
