## 2. Calculus: The Engine of Learning

If Linear Algebra provides the structures, Calculus provides the movement. To train a model, we need to know how changing a weight $w$ affects our error $L$. This is the **Derivative**.

### Key Concepts
- **The Derivative (Gradient):** The rate of change. In multi-variable calculus, the 'Gradient' $\nabla$ is a vector of partial derivatives.
- **The Chain Rule:** This is the most important rule in Deep Learning. It allows us to calculate the derivative of a composite function. If $y = f(g(x))$, then:
  $$\frac{dy}{dx} = \frac{dy}{dg} \cdot \frac{dg}{dx}$$
  In neural networks, this allows us to propagate the error from the output layer back to the very first weight (Backpropagation).

### Power Rule Example
If $f(x) = x^2$, the derivative is $f'(x) = 2x$. If we want to minimize $f(x)$, the derivative tells us which direction to move $x$ to decrease the value.

## 4. Optimization: Finding the Global Minimum

Training a neural network is essentially an optimization problem. We have a **Loss Function** $L$ that measures the error, and we want to find the weights $W$ that minimize this loss.

### Gradient Descent
This is the bread and butter of optimization. We calculate the gradient of the loss with respect to the weights and take a small step in the opposite direction:
$$W_{new} = W_{old} - \eta \cdot \nabla_W L$$
where $\eta$ (eta) is the **Learning Rate**.

**Reasoning**:
I will write a Python script using NumPy to simulate a simple gradient descent process. This will include defining a loss function (quadratic), its gradient, and then performing iterative updates to find the minimum.



````python
import numpy as np

# 1. Define a simple Loss Function: L(w) = w^2
def loss_function(w):
    return w**2

# 2. Define the Gradient: dL/dw = 2w
def gradient(w):
    return 2 * w

# 3. Hyperparameters
learning_rate = 0.1
iterations = 20
weight = 10.0  # Initial starting point

print(f"Starting optimization at w = {weight}\n")

# 4. Optimization Loop (Gradient Descent)
for i in range(iterations):
    # Calculate gradient
    grad = gradient(weight)

    # Update weight: w = w - lr * grad
    weight = weight - (learning_rate * grad)

    # Calculate current loss
    current_loss = loss_function(weight)

    if i % 2 == 0:
        print(f"Iteration {i+1}: w = {weight:.4f}, Loss = {current_loss:.4f}")

print(f"\nFinal weight after {iterations} iterations: {weight:.4f}")
````

## 5. Neural Network Mechanics: The Artificial Neuron

A neural network is simply a collection of nested functions. At the core is the **Neuron** (or Unit).

### The Anatomy of a Neuron
A neuron takes inputs ($x$), multiplies them by weights ($w$), adds a bias ($b$), and then passes the result through a non-linear **Activation Function** ($σ$):
$$z = ∑(w_i x_i) + b$$
$$a = σ(z)$$

### Why Non-Linearity?
Without activation functions, a neural network—no matter how many layers it has—would just be a single linear transformation (a simple regression). Non-linear functions like **ReLU** or **Sigmoid** allow the network to learn complex, non-linear patterns in data.

### Backpropagation: The Chain Rule in Action
To update weights, we need the gradient of the Loss ($L$) with respect to each weight ($w$). Using the chain rule:
$$\frac{\partial L}{\partial w} = \frac{\partial L}{\partial a} \cdot \frac{\partial a}{\partial z} \cdot \frac{\partial z}{\partial w}$$
This process of flowing gradients backward from the output to the input is called **Backpropagation**.

**Reasoning**:
I will implement the Sigmoid and ReLU activation functions and their derivatives. These are essential for the backpropagation process in a neural network.



````python
import numpy as np
import matplotlib.pyplot as plt

# 1. Sigmoid Activation Function and its Derivative
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    # d/dx sigmoid(x) = sigmoid(x) * (1 - sigmoid(x))
    s = sigmoid(x)
    return s * (1 - s)

# 2. ReLU (Rectified Linear Unit) and its Derivative
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return np.where(x > 0, 1, 0)

# Visualize the functions
x = np.linspace(-5, 5, 100)

plt.figure(figsize=(12, 5))

# Plot Sigmoid
plt.subplot(1, 2, 1)
plt.plot(x, sigmoid(x), label='Sigmoid', color='blue', lw=2)
plt.plot(x, sigmoid_derivative(x), label='Derivative', linestyle='--', color='darkblue')
plt.title("Sigmoid & Derivative")
plt.grid(True)
plt.legend()

# Plot ReLU
plt.subplot(1, 2, 2)
plt.plot(x, relu(x), label='ReLU', color='red', lw=2)
plt.plot(x, relu_derivative(x), label='Derivative', linestyle='--', color='darkred')
plt.title("ReLU & Derivative")
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()

print("Activation functions defined and visualized successfully.")
````

## 6. Entering the Modern Era: PyTorch & Autograd

While understanding the manual math in NumPy is crucial for intuition, modern Deep Learning relies on frameworks like **PyTorch** to handle the heavy lifting.

### Why PyTorch?
1. **GPU Acceleration:** PyTorch Tensors can be moved to GPUs (NVIDIA/Metal) to perform massive parallel computations, which is significantly faster than CPU-bound NumPy.
2. **Autograd (Automatic Differentiation):** This is the killer feature. You no longer need to manually derive and code derivatives like `sigmoid_derivative`. PyTorch tracks every operation and computes the gradients for you automatically.
3. **Dynamic Computation Graphs:** PyTorch builds the 'graph' of operations on the fly, making it flexible for complex architectures.

### Tensors vs. NumPy Arrays
A PyTorch `Tensor` is very similar to a NumPy `ndarray`, but with added metadata to support the computation of gradients (`requires_grad=True`).

| Feature | NumPy | PyTorch |
| :--- | :--- | :--- |
| **Device** | CPU Only | CPU, GPU, TPU |
| **Gradients** | Manual implementation | Automatic (Autograd) |
| **Usage** | General Data Science | Deep Learning / Research |

## 7. Common Pitfalls & Best Practices

Transitioning from theory to code often introduces subtle bugs. Here are the most common hurdles for beginners:

### 1. Numerical Instability
Computers have finite precision. Operations like `log(0)` or `exp(large_number)` can result in `NaN` (Not a Number) or `Inf`.
*   **Fix:** Add a small epsilon ($̕ = 1e-7$) to log arguments: `np.log(x + 1e-7)`.

### 2. Dimension Mismatch & Broadcasting
NumPy and PyTorch 'broadcast' arrays of different shapes, which can lead to silent errors where the code runs but the math is wrong.
*   **Tip:** Always check `.shape` before performing matrix operations.

### 3. Vanishing Gradients
Using Sigmoid in very deep networks can cause gradients to become so small that the early layers stop learning.
*   **Fix:** Prefer **ReLU** for hidden layers as it maintains a constant gradient for positive inputs.

**Reasoning**:
I will begin the mini-project by implementing the core logic of a simple MLP using NumPy. This includes the initialization, forward pass, and the training loop for the XOR problem.



````python
import numpy as np

# 1. Dataset: XOR problem (Non-linear)
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([[0], [1], [1], [0]])

# 2. Parameters Initialization
input_size = 2
hidden_size = 4
output_size = 1
np.random.seed(42)

W1 = np.random.randn(input_size, hidden_size)
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size)
b2 = np.zeros((1, output_size))

learning_rate = 0.1
epochs = 10000

# 3. Training Loop
for epoch in range(epochs):
    # Forward Pass
    z1 = np.dot(X, W1) + b1
    a1 = np.tanh(z1) # Hidden layer activation
    z2 = np.dot(a1, W2) + b2
    a2 = 1 / (1 + np.exp(-z2)) # Output layer activation (Sigmoid)

    # Compute Loss (MSE)
    loss = np.mean((a2 - y)**2)

    # Backward Pass (Chain Rule)
    dz2 = (a2 - y) * (a2 * (1 - a2))
    dW2 = np.dot(a1.T, dz2)
    db2 = np.sum(dz2, axis=0, keepdims=True)

    da1 = np.dot(dz2, W2.T)
    dz1 = da1 * (1 - np.tanh(z1)**2) # Derivative of tanh
    dW1 = np.dot(X.T, dz1)
    db1 = np.sum(dz1, axis=0, keepdims=True)

    # Update Weights
    W2 -= learning_rate * dW2
    b2 -= learning_rate * db2
    W1 -= learning_rate * dW1
    b1 -= learning_rate * db1

    if epoch % 2000 == 0:
        print(f"Epoch {epoch}, Loss: {loss:.4f}")

print("\nFinal Predictions:")
print(a2.round(2))
````

## 1. The Problem: Sparse Representation (One-Hot Encoding)
Historically, words were represented as **One-Hot Vectors**. If your vocabulary has 10,000 words, the word "cat" is a vector of 9,999 zeros and a single '1' at a specific index.

**The Critical Flaws:**
1. **High Dimensionality:** As vocabulary grows (e.g., 100k+ words), these vectors become computationally expensive.
2. **Orthogonality:** Every one-hot vector is perpendicular to every other. Mathematically, the distance between "cat" and "dog" is the same as the distance between "cat" and "skyscraper". There is **zero semantic relationship** captured.

## 2. The Solution: Dense Embeddings
Embeddings represent words as **low-dimensional, dense, real-valued vectors** (e.g., 300 dimensions instead of 50,000). In this space, words with similar meanings are forced to cluster together, allowing us to perform 'word arithmetic':

$$\text{Vector("King")} - \text{Vector("Man")} + \text{Vector("Woman")} \approx \text{Vector("Queen")}$$


## 3. The Distributional Hypothesis

How do we learn these dense vectors? The foundation is the **Distributional Hypothesis**:

> *"You shall know a word by the company it keeps"* (Firth, 1957).

If the words "coffee" and "tea" often appear in similar contexts (e.g., "I drink ___ in the morning"), a model should learn that they are related. Algorithms like **Word2Vec** operationalize this by training a simple neural network to predict a word from its neighbors (**CBOW**) or predict the neighbors from a word (**Skip-gram**).

## 4. Measuring Meaning: Vector Similarity

Once words are in a vector space, we need a way to measure how 'close' they are.

### Cosine Similarity
This is the most common metric in NLP. It measures the cosine of the angle between two vectors. It focuses on the **direction** rather than the magnitude.

$$\text{Cosine Similarity}(A, B) = \frac{A \cdot B}{\|A\| \|B\|}$$

- **1.0:** Vectors point in the exact same direction (identical meaning).
- **0.0:** Vectors are orthogonal (unrelated).
- **-1.0:** Vectors point in opposite directions (rare in NLP embeddings).

### Euclidean Distance (L2)
Measures the straight-line distance between two points in space.

$$d(A, B) = \sqrt{\sum (A_i - B_i)^2}$$

While Cosine Similarity is scale-invariant, Euclidean distance is sensitive to the magnitude of the vectors, which can be influenced by word frequency during training.

**Reasoning**:
I will start by implementing the NumPy-based embedding lookup and the cosine similarity function to demonstrate the core mechanics of embeddings manually.



````python
import numpy as np

# 1. Manual Embedding Lookup in NumPy
# Imagine a tiny vocabulary: ['king', 'queen', 'apple', 'orange']
vocab_size = 4
embedding_dim = 2

# Initialize a weight matrix (The Embedding Layer)
# Each row corresponds to a word vector
np.random.seed(42)
W_embed = np.random.randn(vocab_size, embedding_dim)

def get_embedding(word_idx, weight_matrix):
    # Indexing into the matrix is equivalent to multiplying a one-hot vector by the weights
    return weight_matrix[word_idx]

# 2. Similarity Measure
def cosine_similarity(v1, v2):
    dot_product = np.dot(v1, v2)
    norm_v1 = np.linalg.norm(v1)
    norm_v2 = np.linalg.norm(v2)
    return dot_product / (norm_v1 * norm_v2)

# Example usage
king_vec = get_embedding(0, W_embed)
queen_vec = get_embedding(1, W_embed)
apple_vec = get_embedding(2, W_embed)

sim_king_queen = cosine_similarity(king_vec, queen_vec)
sim_king_apple = cosine_similarity(king_vec, apple_vec)

print(f"Embedding Matrix:\n{W_embed}\n")
print(f"Similarity (King, Queen): {sim_king_queen:.4f}")
print(f"Similarity (King, Apple): {sim_king_apple:.4f}")
````

````python
import matplotlib.pyplot as plt
import torch
import torch.nn as nn

# --- 1. Visualizing the Embedding Space ---
words = ['king', 'queen', 'apple', 'orange']
x_coords = W_embed[:, 0]
y_coords = W_embed[:, 1]

plt.figure(figsize=(8, 6))
plt.scatter(x_coords, y_coords, color='purple', s=100)

# Annotate the points
for i, word in enumerate(words):
    plt.annotate(word, (x_coords[i], y_coords[i]), xytext=(5, 5), textcoords='offset points', fontsize=12)

plt.title("2D Visualization of Word Embeddings (NumPy)")
plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.grid(True, linestyle='--', alpha=0.6)
plt.axhline(0, color='black',linewidth=0.5)
plt.axvline(0, color='black',linewidth=0.5)
plt.show()

# --- 2. The PyTorch Equivalent: nn.Embedding ---
# In PyTorch, nn.Embedding is essentially a lookup table that stores embeddings of a fixed dictionary and size.

# Define the layer
embedding_layer = nn.Embedding(num_embeddings=vocab_size, embedding_dim=embedding_dim)

# Let's manually set the weights to match our NumPy matrix for comparison
with torch.no_grad():
    embedding_layer.weight.copy_(torch.from_numpy(W_embed))

# Input to nn.Embedding must be LongTensor (indices)
indices = torch.tensor([0, 1]) # king, queen

# Perform lookup
output_embeddings = embedding_layer(indices)

print("PyTorch nn.Embedding lookup for [king, queen]:")
print(output_embeddings)
print(f"\nShape of output: {output_embeddings.shape} (Batch Size x Embedding Dim)")
````

## 5. Tokenization: Breaking Down the World

Before we can look up an embedding, we must break text into discrete units called **tokens**. The way we do this has evolved significantly.

### The Evolution of Tokenization

1. **Word-level:** Splitting by whitespace.
   - *Problem:* Huge vocabulary sizes and the **Out-of-Vocabulary (OOV)** problem. If the model sees 'playing' but only knows 'play', it fails.
2. **Character-level:** Splitting into individual letters.
   - *Problem:* Loses semantic meaning at each step and results in very long sequences.
3. **Sub-word Tokenization (The Modern Standard):** Splitting words into frequent chunks (e.g., 'un' + 'happi' + 'ness').

### Modern Algorithms

| Algorithm | Logic | Used In |
| :--- | :--- | :--- |
| **BPE (Byte Pair Encoding)** | Iteratively merges the most frequent pair of adjacent characters/sequences. | GPT-2, GPT-3, Llama |
| **WordPiece** | Similar to BPE but merges pairs that maximize the likelihood of the training data. | BERT |
| **SentencePiece** | Treats the input as a raw stream (including spaces as tokens), avoiding language-specific pre-tokenizers. | T5, Llama 2 |

## 6. Common Pitfalls in Embedding Layers

### 1. Embedding Collapse
During training, if the learning rate is too high or the regularization is poor, word vectors can collapse into a single point or a very narrow cone, losing their ability to distinguish meanings.

### 2. The OOV (Out-of-Vocabulary) Trap
In word-level models, unseen words are mapped to a single `[UNK]` (Unknown) token. If your `[UNK]` rate is too high, your model is essentially 'blind' to new data. Sub-word tokenization solves this by breaking unknown words into known sub-pieces.

### 3. Static vs. Contextual Embeddings
Word2Vec/GloVe are **static**. The word 'Bank' has the same vector in 'River bank' and 'Bank account'. Modern models (Transformers) use **Contextual Embeddings** where the vector changes based on the surrounding words.

## 1. The Problem: The Bag-of-Words Limitation

Imagine the following two sentences:
1. *"The dog bit the man."*
2. *"The man bit the dog."*

If we only use the word embeddings we built in Phase 1, the **sum** or **average** of the vectors in both sentences would be identical. To a standard embedding-based model, these sentences are a "Bag of Words"—it knows which words are present, but it has no idea who did what to whom.

### Order is Information
In natural language, the position of a word determines its grammatical role and its relationship to other words. To solve this, we must inject **positional information** into our embeddings so that the vector for "dog" at position 1 is mathematically distinct from the vector for "dog" at position 5.

### Adding vs. Concatenating
The most common approach (standardized by the Transformer architecture) is to **add** a positional vector $PE$ to the word embedding $E$:
$$X = E + PE$$

This requires the positional encoding to have the same dimensionality as the embedding and to follow a specific mathematical structure that the model can learn to 'decode' during training.

## Sinusoidal Positional Encoding Theory & Math

### Subtask:
Detail the mathematical intuition and formulas behind fixed sinusoidal positional encodings.


## 2. Sinusoidal Positional Encoding: The Mathematical Intuition

How do we create a vector that represents 'Position 3' in a way that the model can understand its relationship to 'Position 4' or 'Position 10'?

In the seminal paper *"Attention Is All You Need"*, the authors proposed using **fixed sinusoidal functions** of different frequencies.

### The Formulas
For a token at position $pos$ and a dimension $i$ (where $d_{model}$ is the total embedding dimension), the encoding is calculated as:

$$PE_{(pos, 2i)} = \sin\left(\frac{pos}{10000^{2i/d_{model}}}\right)$$
$$PE_{(pos, 2i+1)} = \cos\left(\frac{pos}{10000^{2i/d_{model}}}\right)$$

### Why Sines and Cosines?
1. **Bounded Values:** The values are always between -1 and 1, keeping the added signal stable.
2. **Relative Symmetry:** For any fixed offset $k$, $PE_{pos+k}$ can be represented as a linear function of $PE_{pos}$. This theoretically allows the model to easily learn to attend by relative positions.
3. **Unique Fingerprints:** By using a range of frequencies (from high frequency in early dimensions to low frequency in later dimensions), every position index gets a unique multi-dimensional 'signature'.
4. **Extrapolation:** Because these are continuous functions, the model can potentially handle sequences longer than those seen during training (though this is limited in practice).

## Manual NumPy Implementation of Sinusoidal Encodings

### Subtask:
Implement the sinusoidal positional encoding formula from scratch using NumPy and visualize the resulting matrix.


**Reasoning**:
I will implement the sinusoidal positional encoding function using NumPy and then visualize the encoding matrix with a heatmap using Matplotlib and Seaborn.



````python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def get_sinusoidal_embeddings(max_seq_len, d_model):
    # Initialize the Positional Encoding matrix
    pe = np.zeros((max_seq_len, d_model))

    # Calculate the position indices (0 to max_seq_len-1)
    position = np.arange(0, max_seq_len)[:, np.newaxis]

    # Calculate the division term for the frequencies
    # div_term = 10000^(2i / d_model)
    div_term = np.exp(np.arange(0, d_model, 2) * -(np.log(10000.0) / d_model))

    # Apply sine to even indices (2i)
    pe[:, 0::2] = np.sin(position * div_term)

    # Apply cosine to odd indices (2i+1)
    pe[:, 1::2] = np.cos(position * div_term)

    return pe

# Generate a PE matrix for visualization
seq_len = 50
d_model = 128
pe_matrix = get_sinusoidal_embeddings(seq_len, d_model)

# Visualize using a Heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(pe_matrix, cmap='RdBu', center=0)
plt.title(f"Sinusoidal Positional Encoding Heatmap ({seq_len}x{d_model})")
plt.xlabel("Embedding Dimension")
plt.ylabel("Sequence Position")
plt.show()

print(f"PE Matrix Shape: {pe_matrix.shape}")
````

## Modern Evolution: RoPE and ALiBi

### Subtask:
Discuss the evolution of positional encodings in modern LLMs, including Rotary Positional Embeddings (RoPE) and Attention with Linear Biases (ALiBi).


## 3. Modern Evolution: Beyond Absolute Positions

While the original sinusoidal encodings were revolutionary, they are **absolute**. This means each position is mapped to a fixed vector regardless of the other tokens in the sequence. Modern Large Language Models (LLMs) have moved toward **Relative Positional Encodings** to improve performance and sequence length extrapolation.

### Rotary Positional Embeddings (RoPE)
Used in models like **Llama**, **PaLM**, and **Mistral**, RoPE represents relative position by rotating the embedding vectors in a 2D plane. Instead of adding a vector, it applies a rotation matrix based on the token's position.

- **How it works:** It pairs up dimensions of the embedding and rotates them by an angle $\theta$ that depends on the position index.
- **Key Advantage:** The dot product between two tokens (their attention score) becomes a function of their **relative distance** rather than their absolute positions.

### Attention with Linear Biases (ALiBi)
Used in models like **MPT** and **Bloom**, ALiBi doesn't add anything to the embeddings at all. Instead, it injects positional information directly into the **Attention Mechanism**.

- **How it works:** It subtracts a constant value (a penalty) from the attention scores, where the penalty increases linearly with the distance between tokens.
- **Key Advantage:** It allows for incredible **length extrapolation**, meaning a model trained on 2,048 tokens can often perform reasonably well on 8,000+ tokens.

### Comparison Table

| Feature | Sinusoidal (Original) | RoPE (Rotary) | ALiBi (Bias-based) |
| :--- | :--- | :--- | :--- |
| **Mechanism** | Additive Vector | Multiplicative Rotation | Attention Penalty |
| **Type** | Absolute | Relative | Relative |
| **Extrapolation** | Poor | Moderate | Excellent |
| **Used In** | Transformer-Base | Llama, Mistral | MPT, Falcon |

## PyTorch Implementation and Layer Integration

### Subtask:
Implement a reusable PositionalEncoding module in PyTorch and demonstrate its integration with an embedding layer.


**Reasoning**:
I will now implement the PyTorch PositionalEncoding class as described in the subtask instructions. Using `register_buffer` is best practice for fixed tensors that should be moved to the GPU with the model but not updated by gradients.



````python
import torch
import torch.nn as nn
import math

class PositionalEncoding(nn.Module):
    def __init__(self, d_model, max_len=5000):
        super(PositionalEncoding, self).__init__()

        # Create a matrix of [max_len, d_model] for positional encodings
        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-math.log(10000.0) / d_model))

        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)

        # Add a batch dimension: [1, max_len, d_model]
        pe = pe.unsqueeze(0)

        # register_buffer ensures pe is saved with the model but not trained
        self.register_buffer('pe', pe)

    def forward(self, x):
        # x shape: [batch_size, seq_len, d_model]
        # We add the positional encoding to the input
        x = x + self.pe[:, :x.size(1), :]
        return x

# --- Demonstration ---
batch_size = 2
seq_len = 10
d_model = 16

# 1. Simulate an Embedding layer output
dummy_embeddings = torch.randn(batch_size, seq_len, d_model)

# 2. Instantiate and apply Positional Encoding
pos_encoder = PositionalEncoding(d_model=d_model, max_len=20)
output = pos_encoder(dummy_embeddings)

print(f"Input shape: {dummy_embeddings.shape}")
print(f"Output shape: {output.shape}")
print("\nExample output for first token in batch (first 5 dims):")
print(output[0, 0, :5])
print("PyTorch PositionalEncoding module implemented successfully.")
````

````python
import numpy as np
import matplotlib.pyplot as plt

# 1. Cosine Similarity Function
def cosine_sim(v1, v2):
    return np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))

# 2. Setup
d_model = 128
seq_len = 100

# Create a constant 'word embedding' for the word 'apple'
# We'll use a random but fixed vector
np.random.seed(42)
apple_embedding = np.random.randn(d_model) * 0.1

# 3. Get Positional Encodings (using the function defined in earlier cells)
# We'll re-calculate here for self-containment
pe = get_sinusoidal_embeddings(seq_len, d_model)

# 4. Create Position-Aware Embeddings
# X = E + PE
pos_aware_embeddings = apple_embedding + pe

# 5. Calculate Similarity from the first token (index 0) to all others
similarities = []
reference_vec = pos_aware_embeddings[0]

for i in range(seq_len):
    sim = cosine_sim(reference_vec, pos_aware_embeddings[i])
    similarities.append(sim)

# 6. Visualization
plt.figure(figsize=(10, 6))
plt.plot(range(seq_len), similarities, color='teal', linewidth=2)
plt.fill_between(range(seq_len), similarities, alpha=0.1, color='teal')

plt.title("Mini Project: Similarity Decay over Position Distance", fontsize=14)
plt.xlabel("Distance from First Token (Position Index)", fontsize=12)
plt.ylabel("Cosine Similarity with First Token", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.axhline(y=similarities[1], color='red', linestyle=':', label='Neighbor Similarity')
plt.legend()
plt.show()

print(f"Similarity with self (pos 0): {similarities[0]:.4f}")
print(f"Similarity with next neighbor (pos 1): {similarities[1]:.4f}")
print(f"Similarity with last token (pos {seq_len-1}): {similarities[-1]:.4f}")
````

## Self-Attention Intuition: The QKV Framework

## 2. Self-Attention Intuition: The QKV Framework

How does a model decide which words are important to one another? To do this, it uses a system inspired by **Information Retrieval**, known as the **Query-Key-Value (QKV)** framework.

Imagine you are in a massive library looking for a specific piece of information.

### The Library Analogy
1.  **Query (Q):** This is the **search term** you type into the library database. It represents what the current word is 'looking for' in the rest of the sentence (e.g., 'Which noun does this adjective describe?').
2.  **Key (K):** This is the **label on the spine** of every book in the library. Every word in the sentence has a 'key' that describes what kind of information it contains.
3.  **Value (V):** This is the **content inside the book**. Once you match your Query with a Key, you extract the 'Value' (the actual semantic information) to update your understanding.

### In the Context of a Sentence
In a Transformer, every single word vector is transformed into three different vectors via learned weight matrices ($W_Q, W_K, W_V$):

*   **Query ($q_i$):** \"I am the word 'Barking'. I am looking for the subject that is performing this action.\"
*   **Key ($k_j$):** \"I am the word 'Dog'. My label says I am a canine/animal subject.\"
*   **Value ($v_j$):** \"Here is the actual information about the 'Dog' to help you understand the context of 'Barking'.\"

### The Interaction
We compare the **Query** of word $A$ with the **Keys** of all other words ($B, C, D$). The better the match, the more **Attention** word $A$ pays to word $B$. We then take a weighted sum of the **Values** based on those attention scores.

**Result:** The word 'Barking' is no longer just a static embedding; it is now 'Barking' *contextualized* by the 'Dog'."

## Mathematical Foundation: Scaled Dot-Product Attention

### Subtask:
Derive and explain the Scaled Dot-Product Attention formula, focusing on the mathematical role of each component.


## 3. Mathematical Foundation: Scaled Dot-Product Attention

Having built the intuition, let's translate the library analogy into linear algebra. The mechanism used in the original Transformer is called **Scaled Dot-Product Attention**.

### The Equation
Given a set of Queries ($Q$), Keys ($K$), and Values ($V$), the attention is calculated as follows:

$$\text{Attention}(Q, K, V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d_k}}\right)V$$

### Breaking Down the Components

#### 1. The Dot Product ($QK^T$)
To find the similarity between a Query and all Keys, we compute their dot product. In matrix form, $QK^T$ calculates every pair-wise similarity between queries and keys.
- A **high positive value** means the query and key are highly aligned (strong attention).
- A **value near zero** means they are unrelated.

#### 2. The Scaling Factor ($\frac{1}{\sqrt{d_k}}$)
This is the most critical technical detail. As the dimension of the keys ($d_k$) grows, the magnitude of the dot products tends to grow very large.

**The Problem:** Large input values to the `softmax` function push it into regions where gradients are extremely small (the 'flat' parts of the curve). This leads to **vanishing gradients**, effectively stopping the model from learning.

**The Solution:** By dividing by $\sqrt{d_k}$, we scale the variance of the dot products back to approximately 1, ensuring stable gradients and efficient training.

#### 3. The Softmax
We apply the softmax function row-wise. This turns the raw similarity scores into a **probability distribution**. Every row will now sum to 1.0, representing the percentage of 'attention' the token should pay to every other token.

#### 4. The Weighted Sum ($...V$)
Finally, we multiply these attention weights by the **Value** matrix. This produces a new vector that is a combination of the most relevant values in the sequence.

**Result:** We have successfully moved from a sequence of independent tokens to a sequence of **context-aware** representations.

## NumPy Implementation: Manual Attention

### Subtask:
Implement a raw NumPy function for Scaled Dot-Product Attention, including softmax normalization and the weighted sum of values.


**Reasoning**:
I will implement the `scaled_dot_product_attention` function in NumPy as specified in the instructions. This will include a simple softmax helper function to ensure the code is self-contained.



````python
import numpy as np

def softmax(x):
    """Compute softmax values for each sets of scores in x."""
    e_x = np.exp(x - np.max(x, axis=-1, keepdims=True))
    return e_x / e_x.sum(axis=-1, keepdims=True)

def scaled_dot_product_attention(Q, K, V):
    """
    Compute Scaled Dot-Product Attention.
    """
    d_k = Q.shape[-1]

    # 1. Dot Product (Similarity scores)
    scores = np.matmul(Q, K.T)

    # 2. Scale to prevent gradient saturation
    scaled_scores = scores / np.sqrt(d_k)

    # 3. Softmax to get Attention Weights
    attention_weights = softmax(scaled_scores)

    # 4. Weighted Sum of Values
    output = np.matmul(attention_weights, V)

    return output, attention_weights

# --- Demonstration ---
# Imagine a sequence of 3 tokens, each with an embedding dim of 4
np.random.seed(42)
seq_len = 3
d_k = 4

Q = np.random.randn(seq_len, d_k)
K = np.random.randn(seq_len, d_k)
V = np.random.randn(seq_len, d_k)

context_vectors, weights = scaled_dot_product_attention(Q, K, V)

print("Attention Weights (Sum of each row should be 1.0):")
print(weights)
print(f"\nWeights sum check: {weights.sum(axis=-1)}")

print("\nContextualized Output (First token):")
print(context_vectors[0])

print("\nNumPy Attention implementation complete.")
````

## Multi-Head Attention (MHA) Theory

### Subtask:
Explain the motivation and mechanism of Multi-Head Attention, detailing the splitting, concatenation, and projection of attention heads.


## 4. Multi-Head Attention (MHA): Attending to Multiple Subspaces

While Scaled Dot-Product Attention is powerful, using only one "head" limits the model's ability to understand complex relationships. For example, in the sentence *"The cat sat on the mat because it was tired,"* the word "it" needs to attend to "cat" to resolve the reference, but it might also need to attend to "tired" to understand the cause. A single attention head would have to average these signals, potentially losing detail.

### The Solution: Parallel Heads
Instead of performing a single attention function, we project the Queries, Keys, and Values $h$ times with different, learned linear projections. This allows the model to jointly attend to information from different representation subspaces at different positions.

### The Process
1. **Linear Projection:** Before attention, $Q, K,$ and $V$ are projected using learned matrices $W_Q, W_K, W_V$.
2. **Splitting:** The projected vectors are split into $h$ heads. If our model dimension $d_{model}$ is 512 and we have 8 heads, each head processes a subspace of dimension $d_k = 512 / 8 = 64$.
3. **Parallel Attention:** Each head performs Scaled Dot-Product Attention independently.
4. **Concatenation:** The outputs from all heads are concatenated together, restoring the original $d_{model}$ dimension.
5. **Final Projection:** A final weight matrix $W_O$ is applied to the concatenated result to ensure the heads can communicate with each other.

### Why MHA Matters
- **Specialization:** One head might focus on grammar, another on semantic meaning, and another on local context (neighboring words).
- **Robustness:** By averaging the results of multiple independent searches, the model becomes more robust to noise in any single representation subspace.

## Masked Attention: The Causal Constraint

### Subtask:
Explain the concept of 'Causal Masking' or 'Look-ahead Masking' in the context of autoregressive models.


## 5. Masked Attention: The Causal Constraint

In models like BERT, every token can see every other token. This is fine for understanding a static sentence. However, for models that **generate** text (like GPT-4), we have a problem. During training, the model is given the entire sentence at once, but we want it to learn to predict the *next* word based only on *past* words.

### The Problem: 'Cheating' in Autoregression
If we let a decoder-only model use standard self-attention, the word at position 2 could look at the word at position 5 to predict itself. This is effectively 'cheating' by looking at the future. To prevent this, we use **Causal Masking** (also known as a Look-ahead mask).

### The Mechanism: Numerical Silence
The mask is a lower-triangular matrix of zeros and negative infinities ($-\infty$). Before we apply the softmax to our attention scores, we add this mask:

1. **Scores Matrix:** Calculate $QK^T / \sqrt{d_k}$.
2. **Apply Mask:** For all indices where $j > i$ (future tokens), we set the score to $-\infty$.
3. **Softmax:** Because $e^{-\infty} = 0$, the softmax function assigns exactly **zero probability** to those future tokens.

### The Masking Grid
For a 4-token sequence, the mask ensures the following visibility:
- **Token 1** can only see **Token 1**.
- **Token 2** can see **Token 1, 2**.
- **Token 3** can see **Token 1, 2, 3**.
- **Token 4** can see **Token 1, 2, 3, 4**.

This forces the model to build representations based solely on the prefix, mimicking the way it will actually generate text during inference.

## Modern Optimizations: MQA and GQA

### Subtask:
Explain the architectural shifts from standard Multi-Head Attention to Multi-Query and Grouped-Query Attention used in modern LLMs.


## 6. Modern Optimizations: MQA and GQA

Standard Multi-Head Attention (MHA) is computationally expensive, especially during **inference** (text generation). As we generate tokens one by one, we store the Keys and Values for all previous tokens in a **KV-Cache**. With MHA, this cache grows massive, often becoming the primary bottleneck for memory bandwidth.

To solve this, modern LLMs like Llama 3, Mistral, and Falcon use optimized attention variants.

### Multi-Query Attention (MQA)
In MQA, we keep multiple **Query** heads but use only **one single Key head** and **one single Value head** shared across all queries.
- **Pros:** Significantly reduces memory bandwidth requirements and the size of the KV-cache.
- **Cons:** Can lead to a slight drop in model quality because the model has less 'memory' to attend to.

### Grouped-Query Attention (GQA)
GQA is the 'Goldilocks' solution used in Llama 3. It groups the Query heads into $N$ groups, and each group shares a single Key and Value head.
- **Mechanism:** If you have 8 Query heads and 2 groups, heads 1-4 share one KV pair, and heads 5-8 share another.
- **Pros:** It provides almost the same speed and memory benefits as MQA but maintains the high performance and quality of MHA.

### Comparison Table

| Feature | Multi-Head (MHA) | Multi-Query (MQA) | Grouped-Query (GQA) |
| :--- | :--- | :--- | :--- |
| **Query Heads** | $H$ | $H$ | $H$ |
| **Key/Value Heads** | $H$ | 1 | $G$ (where $1 < G < H$) |
| **Memory Usage** | High | Very Low | Low |
| **Model Quality** | Highest | Lowest | Near-MHA |
| **Used In** | Original Transformer, BERT | Falcon (original), PaLM | Llama 3, Mistral |

## Advanced Speed: FlashAttention Overview

### Subtask:
Explain the IO-awareness and tiling concepts behind FlashAttention that allow for faster processing of long sequences.


## 7. Advanced Speed: FlashAttention Overview

While MQA and GQA change the *architecture* of the model to save memory, **FlashAttention** is a hardware-level optimization that speeds up the *exact same* mathematical calculation of standard attention.

### The Problem: The 'Memory Wall'
Standard attention is **Memory-Bound**, not Compute-Bound. GPUs are incredibly fast at doing math (FLOPs), but relatively slow at moving data between their different types of memory:
1. **HBM (High Bandwidth Memory):** Large but slow (where the $N \times N$ attention matrix usually sits).
2. **SRAM:** Very small but extremely fast (where the actual computation happens).

In standard attention, we calculate the $N \times N$ scores, write them to HBM, read them back for softmax, write them to HBM again, etc. This constant 'shuffling' of data is what slows us down.

### The Solution: IO-Awareness & Tiling
FlashAttention avoids writing the large $N \times N$ matrix to the slow HBM entirely. It uses two clever tricks:

1.  **Tiling:** It breaks the Query, Key, and Value matrices into small blocks (tiles) that fit entirely inside the fast **SRAM**. It performs all the math for one block (dot product, softmax, weighted sum) before moving to the next.
2.  **Recomputation:** To calculate softmax across tiles, it keeps track of partial softmax normalization statistics. If it needs a value again during the backward pass, it simply re-calculates it rather than reading it from slow memory.

### Why it Matters
- **Speed:** It can be 2-4x faster than standard PyTorch attention.
- **Memory Efficiency:** It reduces memory usage from quadratic $O(N^2)$ to linear $O(N)$ with respect to sequence length for the intermediate steps.
- **Long Context:** FlashAttention is the primary reason modern models can handle contexts of 32k, 128k, or even 1M tokens.

## PyTorch Implementation: Building a MultiHeadAttention Module

### Subtask:
Develop a modular and efficient Multi-Head Attention class using PyTorch's nn.Module.


**Reasoning**:
I will implement the `MultiHeadAttention` class in PyTorch, ensuring it correctly handles the splitting of heads and the scaled dot-product calculation while maintaining modularity.



````python
import torch
import torch.nn as nn
import torch.nn.functional as F
import math

class MultiHeadAttention(nn.Module):
    def __init__(self, d_model, num_heads):
        super(MultiHeadAttention, self).__init__()
        assert d_model % num_heads == 0, "d_model must be divisible by num_heads"

        self.d_model = d_model
        self.num_heads = num_heads
        self.d_k = d_model // num_heads

        # Linear layers for Q, K, V projections
        self.w_q = nn.Linear(d_model, d_model)
        self.w_k = nn.Linear(d_model, d_model)
        self.w_v = nn.Linear(d_model, d_model)

        # Final output projection
        self.w_o = nn.Linear(d_model, d_model)

    def forward(self, q, k, v, mask=None):
        batch_size = q.size(0)

        # 1. Linear Projections
        # (batch, seq_len, d_model) -> (batch, seq_len, num_heads, d_k) -> (batch, num_heads, seq_len, d_k)
        query = self.w_q(q).view(batch_size, -1, self.num_heads, self.d_k).transpose(1, 2)
        key = self.w_k(k).view(batch_size, -1, self.num_heads, self.d_k).transpose(1, 2)
        value = self.w_v(v).view(batch_size, -1, self.num_heads, self.d_k).transpose(1, 2)

        # 2. Scaled Dot-Product Attention
        # (batch, num_heads, seq_len, d_k) @ (batch, num_heads, d_k, seq_len) -> (batch, num_heads, seq_len, seq_len)
        scores = torch.matmul(query, key.transpose(-2, -1)) / math.sqrt(self.d_k)

        if mask is not null:
            scores = scores.masked_fill(mask == 0, -1e9)

        attention_weights = F.softmax(scores, dim=-1)

        # 3. Weighted Sum
        # (batch, num_heads, seq_len, seq_len) @ (batch, num_heads, seq_len, d_k) -> (batch, num_heads, seq_len, d_k)
        context = torch.matmul(attention_weights, value)

        # 4. Concatenation and Final Projection
        # (batch, num_heads, seq_len, d_k) -> (batch, seq_len, d_model)
        context = context.transpose(1, 2).contiguous().view(batch_size, -1, self.d_model)
        output = self.w_o(context)

        return output, attention_weights

# --- Test Implementation ---
batch_size, seq_len, d_model, n_heads = 2, 8, 128, 8
x = torch.randn(batch_size, seq_len, d_model)

mha = MultiHeadAttention(d_model, n_heads)
out, weights = mha(x, x, x)

print(f"Input shape: {x.shape}")
print(f"Output shape: {out.shape}")
print(f"Attention weights shape: {weights.shape}")
print("\nMulti-Head Attention implementation successful.")
````

## Memory and Compute Complexity

### Subtask:
Analyze and explain the O(N^2) complexity of the self-attention mechanism and its implications for sequence length scaling.


## 8. Memory and Compute Complexity: The Quadratic Bottleneck

While Attention allows for perfect parallelism, it comes with a steep price tag in terms of computational resources. This is known as the **Quadratic Bottleneck**.

### The $O(N^2)$ Reality
In self-attention, every token must compare itself to every other token in the sequence. If you have a sequence of length $N$:
1.  **Dot Product Matrix:** Computing $QK^T$ requires calculating $N 	imes N$ similarity scores.
2.  **Softmax & Value Sum:** You must then normalize and aggregate across these $N^2$ elements.

Mathematically, the complexity is $O(N^2 	imes d)$, where $d$ is the embedding dimension.

### Comparison with RNNs
| Feature | RNN / LSTM | Transformer (Attention) |
| :--- | :--- | :--- |
| **Time Complexity** | $O(N 	imes d^2)$ | $O(N^2 	imes d)$ |
| **Parallelization** | No (Sequential) | Yes (Parallel) |
| **Long-range Dependency** | Poor (Vanishing Gradients) | Perfect (Direct Connection) |

### Why does this matter?
- **Memory Explosions:** If you double your input length (e.g., from 1,000 to 2,000 tokens), the attention matrix doesn't double—it becomes **four times larger**.
- **The 'Context Window' limit:** This quadratic growth is the primary reason why early models were limited to 512 or 2,048 tokens. Scaling to 100k+ tokens requires the modern optimizations we discussed earlier, like **FlashAttention**, which manage this memory more efficiently without changing the fundamental $O(N^2)$ nature of the math.

## Common Pitfalls in Attention Implementation

### Subtask:
Identify and explain the most frequent errors encountered when implementing self-attention mechanisms, covering mathematical, logical, and dimensional mistakes.


## 9. Common Pitfalls in Attention Implementation

Implementing the Attention mechanism correctly requires precision. Even small logical errors can lead to models that either fail to train or behave like a 'Bag of Words'.

### 1. Forgetting the Scaling Factor ($1/\sqrt{d_k}$)
This is the most common mathematical error. Without scaling, the dot products grow large in high-dimensional spaces, pushing the softmax function into its flat regions.
- **Consequence:** Gradients become nearly zero (Vanishing Gradients), and the model stops learning entirely.

### 2. Softmax Over the Wrong Axis
In self-attention, the softmax must be applied across the **Key** dimension (usually `dim=-1`) for every Query.
- **Consequence:** If applied across the Query dimension, you are normalizing the wrong relationship, resulting in a probability distribution that doesn't represent attention weights.

### 3. Incorrect Head Splitting
When moving from $d_{model}$ to multiple heads, the tensor must be reshaped and transposed such that the attention calculation happens independently for each head.
- **Consequence:** If you simply reshape without transposing (bringing the `num_heads` dimension forward), your dot product will accidentally mix information across different heads, breaking the multi-head benefit.

### 4. Broadened Attention (Lack of Masking)
Forgetting to apply the Causal Mask in a decoder model allows the current token to see future tokens.
- **Consequence:** The model 'cheats' during training by looking ahead, resulting in excellent training loss but complete failure during real-world inference/generation.

---

### ✅ Checklist for Success
- [ ] Did I divide $QK^T$ by $\sqrt{d_k}$?
- [ ] Is my softmax applied to the last dimension (`dim=-1`) of the scores matrix?
- [ ] Did I transpose my heads to `(batch, heads, seq, d_k)` before the dot product?
- [ ] (For Generative Models) Is my causal mask correctly zeroing out the upper-triangular future tokens?

## 10. Attention Internals: Interview Prep

To master Transformers, you must be able to explain the 'why' behind the architecture. Here are the most common interview questions regarding Phase 3.

### 🟢 Green: Easy (Foundations)
**Q: What is the primary purpose of the 'Value' (V) matrix in self-attention?**
*   **Answer:** While the Query and Key determine **how much** attention to pay to other tokens, the Value matrix contains the **actual information** (the content) of the tokens. The final output of the attention layer is a weighted sum of these Value vectors.

### 🟡 Yellow: Medium (Architecture & Math)
**Q: Why do we use $\sqrt{d_k}$ as a scaling factor? What happens if we omit it?**
*   **Answer:** In high dimensions ($d_k$), the dot product of two vectors can grow very large in magnitude. This pushes the softmax function into regions where the gradient is extremely small (the 'saturation' or 'flat' regions). Without the scaling factor, the model suffers from **vanishing gradients**, making it nearly impossible to train.

**Q: How does Multi-Head Attention (MHA) maintain the same computational cost as Single-Head Attention?**
*   **Answer:** MHA doesn't just add more parameters; it splits the existing $d_{model}$ into $h$ heads. Each head operates on a lower-dimensional subspace ($d_k = d_{model} / h$). The total number of operations remains roughly the same, but the model can now attend to different semantic features (e.g., one head for syntax, one for semantics) in parallel.

### 🔴 Red: Hard (Engineering & Research)
**Q: Explain the memory benefit of Grouped-Query Attention (GQA) over standard Multi-Head Attention during inference.**
*   **Answer:** The primary bottleneck during inference is the size of the **KV-Cache**. In MHA, every Query head has a corresponding Key/Value head. In GQA, multiple Query heads share a single Key/Value head. This significantly reduces the amount of memory that needs to be moved from HBM to SRAM for every new token generated, leading to higher throughput and lower latency.

**Q: FlashAttention calculates the exact same mathematical output as standard attention. How can it be 2-4x faster?**
*   **Answer:** It is an **IO-Aware** algorithm. Standard attention is 'Memory-Bound', meaning the GPU spends most of its time waiting for the large $N \times N$ attention matrix to be read/written from slow HBM. FlashAttention uses **tiling** to keep everything in fast SRAM and **recomputation** (calculating values on the fly during the backward pass instead of storing them) to avoid the HBM bottleneck entirely.

## Mini Project: The 'Contextualizer'

### Subtask:
Build a script demonstrating how a word embedding's representation is altered by its context words after passing through a self-attention layer.


**Reasoning**:
I will implement the 'Contextualizer' mini-project by defining a small vocabulary and two sentences that change the meaning of the word 'bank'. I will then use the MultiHeadAttention module to show how the vector for 'bank' shifts in the embedding space based on its neighbors.



````python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# 1. Setup Mini-Vocabulary and Embeddings
# Let's define a small dimension for visibility
d_model = 16
vocab = {
    "bank": 0, "river": 1, "water": 2, "money": 3, "finance": 4
}

# Create manual embeddings: Bank is neutral, others are distinct
embeddings = nn.Embedding(5, d_model)
with torch.no_grad():
    embeddings.weight.fill_(0.0)
    embeddings.weight[0] = torch.randn(d_model) * 0.1 # Bank (Central)
    embeddings.weight[1] = torch.ones(d_model) * 0.5  # River (Positive Cluster)
    embeddings.weight[2] = torch.ones(d_model) * 0.6  # Water (Positive Cluster)
    embeddings.weight[3] = torch.ones(d_model) * -0.5 # Money (Negative Cluster)
    embeddings.weight[4] = torch.ones(d_model) * -0.6 # Finance (Negative Cluster)

# 2. Define two sequences (Sentences)
sentence_river = torch.tensor([[0, 1, 2]]) # "bank river water"
sentence_money = torch.tensor([[0, 3, 4]]) # "bank money finance"

# 3. Apply Multi-Head Attention
mha = MultiHeadAttention(d_model, num_heads=2)

with torch.no_grad():
    # Get raw embeddings
    raw_bank = embeddings(torch.tensor([0]))

    # Context 1: River bank
    out_river, _ = mha(embeddings(sentence_river), embeddings(sentence_river), embeddings(sentence_river))
    bank_in_river = out_river[0, 0, :] # First token in first sentence

    # Context 2: Money bank
    out_money, _ = mha(embeddings(sentence_money), embeddings(sentence_money), embeddings(sentence_money))
    bank_in_money = out_money[0, 0, :] # First token in second sentence

# 4. Visualize using PCA to 2D
vectors = torch.stack([raw_bank.squeeze(), bank_in_river, bank_in_money]).numpy()
labels = ['Bank (Raw)', 'Bank (River Context)', 'Bank (Money Context)']

pca = PCA(n_components=2)
coords = pca.fit_transform(vectors)

plt.figure(figsize=(10, 6))
plt.scatter(coords[:, 0], coords[:, 1], c=['black', 'blue', 'green'], s=100)

for i, label in enumerate(labels):
    plt.annotate(label, (coords[i, 0], coords[i, 1]), xytext=(5, 5), textcoords='offset points')

plt.title("The Contextualizer: How Attention Shifts Meaning")
plt.grid(True, linestyle='--', alpha=0.6)
plt.axhline(0, color='black', lw=0.5)
plt.axvline(0, color='black', lw=0.5)
plt.show()

print("Notice how the 'Bank' vector (originally central) is pulled in opposite directions depending on its neighbors!")
````

### 📐 Mathematical Justification: The $1/\sqrt{d_k}$ Scaling Factor

In Scaled Dot-Product Attention, we compute the similarity between a Query ($q$) and a Key ($k$) using the dot product: $q \cdot k = \sum_{i=1}^{d_k} q_i k_i$.

#### The Proof of Variance
Assume that the components of $q$ and $k$ are independent random variables with mean 0 and variance 1 ($q_i, k_i \sim \mathcal{N}(0, 1)$).

1. The mean of the product $q_i k_i$ is:
   $$E[q_i k_i] = E[q_i]E[k_i] = 0 \cdot 0 = 0$$
2. The variance of the product $q_i k_i$ is:
   $$Var(q_i k_i) = E[q_i^2 k_i^2] - (E[q_i k_i])^2 = E[q_i^2]E[k_i^2] - 0 = 1 \cdot 1 = 1$$
3. Since the components are independent, the variance of the sum (the dot product) is the sum of the variances:
   $$Var(q \cdot k) = Var\left(\sum_{i=1}^{d_k} q_i k_i\right) = \sum_{i=1}^{d_k} Var(q_i k_i) = d_k$$

#### The Necessity of Scaling
As $d_k$ increases, the variance of the dot product grows linearly. Large values in the dot product push the input to the **Softmax** function into regions where the gradient is extremely small (the 'flat' tails of the exponential). This leads to the **Vanishing Gradient** problem.

By dividing the dot product by $\sqrt{d_k}$, we scale the variance back to 1:
$$Var\left(\frac{q \cdot k}{\sqrt{d_k}}\right) = \frac{1}{d_k} Var(q \cdot k) = \frac{d_k}{d_k} = 1$$

This ensures that the Softmax function operates in a region where gradients are healthy, allowing the model to learn efficiently.

## 2. The Position-wise Feed-Forward Network (FFN)

While Multi-Head Attention allows tokens to communicate with each other, the **Feed-Forward Network (FFN)** allows the model to process the information gathered for each token individually and non-linearly.

### The Architecture
The FFN consists of two linear transformations with a non-linear activation function in between. It follows a specific expansion-contraction pattern:
1.  **Expansion:** The input dimension $d_{model}$ is projected into a much larger latent space (typically $4 \times d_{model}$). This allows the model to map features into a higher-dimensional space where they might be more easily separable.
2.  **Activation:** A non-linearity like **ReLU** or **GELU** is applied.
3.  **Contraction:** The latent representation is projected back down to $d_{model}$.

### The Mathematical Formula
$$\text{FFN}(x) = \max(0, xW_1 + b_1)W_2 + b_2$$

### Why 'Position-wise'?
The exact same linear layers are applied to **every token position** independently. There is no interaction between positions in this layer—that responsibility belongs solely to the Attention mechanism. Think of the FFN as a 'local processor' that refines the meaning of each token based on the context attention just provided.

**Reasoning**:
I will implement the `PositionwiseFeedForward` class in PyTorch, which includes two linear layers and an activation function, following the standard expansion-contraction pattern (d_model -> 4*d_model -> d_model).



````python
import torch
import torch.nn as nn
import torch.nn.functional as F

class PositionwiseFeedForward(nn.Module):
    def __init__(self, d_model, d_ff, dropout=0.1):
        super(PositionwiseFeedForward, self).__init__()
        # First linear layer: Expansion
        self.w_1 = nn.Linear(d_model, d_ff)
        # Second linear layer: Contraction
        self.w_2 = nn.Linear(d_ff, d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # x shape: (batch_size, seq_len, d_model)
        # 1. Expand and apply non-linearity (ReLU)
        x = F.relu(self.w_1(x))
        # 2. Apply dropout for regularization
        x = self.dropout(x)
        # 3. Project back to d_model
        x = self.w_2(x)
        return x

# --- Verification ---
d_model = 512
d_ff = 2048 # Standard 4x expansion
ffn = PositionwiseFeedForward(d_model, d_ff)

dummy_input = torch.randn(1, 10, d_model)
output = ffn(dummy_input)

print(f"Input shape: {dummy_input.shape}")
print(f"Output shape: {output.shape}")
print("FFN implementation verified: Input and Output dimensions are identical.")
````

## 3. Residual Connections & Layer Normalization

To build deep networks, we need more than just layers; we need mechanisms to ensure that information and gradients can flow through the entire stack without degrading.

### Residual (Skip) Connections
Introduced in ResNets, **Residual Connections** allow the input of a layer to skip the transformation and be added directly to the output:
$$y = x + \text{Sublayer}(x)$$

**Why it matters:** In deep networks, gradients can 'vanish' as they are multiplied by small weights during backpropagation. The skip-connection provides a 'highway' for the gradient to flow back to earlier layers unchanged, making it possible to train models with hundreds of layers (like GPT-3).

### Layer Normalization (LayerNorm)
**Layer Normalization** stabilizes the distribution of activations by ensuring they have a mean of 0 and a variance of 1 across the feature dimension ($d_{model}$) for each token.

Unlike Batch Normalization, which depends on other samples in a batch, LayerNorm operates independently on each token position, making it ideal for sequence modeling where lengths vary.

### The 'Add & Norm' Pattern
In the original Transformer paper, the authors used **Post-LayerNorm**:
1. $x' = \text{LayerNorm}(x + \text{Sublayer}(x))$

However, modern state-of-the-art models (Llama, GPT-3) use **Pre-LayerNorm**:
1. $x' = x + \text{Sublayer}(\text{LayerNorm}(x))$

**Key Difference:** Pre-LayerNorm is generally more stable because the identity path ($x$) remains 'clean' (un-normalized) throughout the stack, facilitating even better gradient flow during the early stages of training.

**Reasoning**:
I will implement the SublayerConnection class, which encapsulates the Pre-LayerNorm logic (LayerNorm followed by the sublayer and a residual add). I will also include a visualization of LayerNorm's effect on activation distributions.



````python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
import seaborn as sns

class SublayerConnection(nn.Module):
    """
    A residual connection followed by a layer norm.
    Note for code simplicity the norm is applied first as per modern Pre-LN architectures.
    """
    def __init__(self, size, dropout):
        super(SublayerConnection, self).__init__()
        self.norm = nn.LayerNorm(size)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x, sublayer):
        "Apply residual connection to any sublayer with the same size."
        # Pre-LayerNorm configuration: x + Dropout(Sublayer(LayerNorm(x)))
        return x + self.dropout(sublayer(self.norm(x)))

# --- Visualization of LayerNorm Effect ---
def visualize_layernorm():
    # 1. Create a biased distribution (mean=5, std=10)
    d_model = 512
    raw_data = torch.randn(100, d_model) * 10 + 5

    # 2. Apply LayerNorm
    ln = nn.LayerNorm(d_model)
    normalized_data = ln(raw_data).detach()

    # 3. Plotting
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    sns.histplot(raw_data.flatten().numpy(), color='salmon', kde=True)
    plt.title(f"Before LayerNorm\nMean: {raw_data.mean():.2f}, Std: {raw_data.std():.2f}")

    plt.subplot(1, 2, 2)
    sns.histplot(normalized_data.flatten().numpy(), color='skyblue', kde=True)
    plt.title(f"After LayerNorm\nMean: {normalized_data.mean():.2f}, Std: {normalized_data.std():.2f}")

    plt.tight_layout()
    plt.show()

visualize_layernorm()
print("SublayerConnection (Pre-LN) implemented and LayerNorm stabilization visualized.")
````

````python
class TransformerBlock(nn.Module):
    """An Encoder layer made up of self-attn and feed forward."""
    def __init__(self, d_model, self_attn, feed_forward, dropout):
        super(TransformerBlock, self).__init__()
        self.self_attn = self_attn
        self.feed_forward = feed_forward
        # We need two sublayer connections: one for MHA, one for FFN
        self.sublayer = nn.ModuleList([
            SublayerConnection(d_model, dropout) for _ in range(2)
        ])
        self.size = d_model

    def forward(self, x, mask=None):
        # 1. First Sublayer: Multi-Head Attention
        # We pass a lambda because SublayerConnection expects a function/module that takes one arg
        x = self.sublayer[0](x, lambda x: self.self_attn(x, x, x, mask)[0])

        # 2. Second Sublayer: Feed-Forward Network
        x = self.sublayer[1](x, self.feed_forward)
        return x

print("TransformerBlock (Encoder Layer) class defined successfully.")
````

````python
import torch
import torch.nn as nn
import torch.nn.functional as F
import math

class MultiHeadAttention(nn.Module):
    def __init__(self, d_model, num_heads):
        super(MultiHeadAttention, self).__init__()
        assert d_model % num_heads == 0, "d_model must be divisible by num_heads"
        self.d_model = d_model
        self.num_heads = num_heads
        self.d_k = d_model // num_heads
        self.w_q = nn.Linear(d_model, d_model)
        self.w_k = nn.Linear(d_model, d_model)
        self.w_v = nn.Linear(d_model, d_model)
        self.w_o = nn.Linear(d_model, d_model)

    def forward(self, q, k, v, mask=None):
        batch_size = q.size(0)
        query = self.w_q(q).view(batch_size, -1, self.num_heads, self.d_k).transpose(1, 2)
        key = self.w_k(k).view(batch_size, -1, self.num_heads, self.d_k).transpose(1, 2)
        value = self.w_v(v).view(batch_size, -1, self.num_heads, self.d_k).transpose(1, 2)

        scores = torch.matmul(query, key.transpose(-2, -1)) / math.sqrt(self.d_k)
        if mask is not None:
            scores = scores.masked_fill(mask == 0, -1e9)
        attn = F.softmax(scores, dim=-1)
        context = torch.matmul(attn, value)
        context = context.transpose(1, 2).contiguous().view(batch_size, -1, self.d_model)
        return self.w_o(context)

class PositionwiseFeedForward(nn.Module):
    def __init__(self, d_model, d_ff, dropout=0.1):
        super(PositionwiseFeedForward, self).__init__()
        self.w_1 = nn.Linear(d_model, d_ff)
        self.w_2 = nn.Linear(d_ff, d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        return self.w_2(self.dropout(F.relu(self.w_1(x))))

# Quick validation check
d_model, n_heads, d_ff = 512, 8, 2048
mha = MultiHeadAttention(d_model, n_heads)
ffn = PositionwiseFeedForward(d_model, d_ff)
x = torch.randn(2, 10, d_model)
print(f"MHA Output Shape: {mha(x,x,x).shape}")
print(f"FFN Output Shape: {ffn(x).shape}")
````

## Encoder Implementation

### Subtask:
Implement a modular Encoder class that aggregates N TransformerBlock layers and applies final normalization.


**Reasoning**:
I will implement the `clones` utility and the `Encoder` class in PyTorch, ensuring that the architecture follows the Pre-LayerNorm requirement with a final normalization step.



````python
import torch
import torch.nn as nn
import copy

def clones(module, N):
    """Produce N identical layers."""
    return nn.ModuleList([copy.deepcopy(module) for _ in range(N)])

class Encoder(nn.Module):
    """Core encoder is a stack of N layers."""
    def __init__(self, layer, N):
        super(Encoder, self).__init__()
        self.layers = clones(layer, N)
        self.norm = nn.LayerNorm(layer.size)

    def forward(self, x, mask=None):
        """Pass the input (and mask) through each layer in turn."""
        for layer in self.layers:
            x = layer(x, mask)
        return self.norm(x)

print("Encoder class and clones utility implemented successfully.")
````

## Phase 5: Deep Dive - The Normalization Zoo

Beyond the structural assembly of stacks, the **stability** of a Transformer depends heavily on how we normalize activations. While we used standard `LayerNorm` previously, modern LLMs often use specialized variants to handle deeper stacks and faster convergence.

### 1. Why Normalize?
As signals pass through many layers, their variance can either explode or vanish. Normalization ensures that activations stay within a healthy numerical range (typically mean 0, variance 1, or just unit variance), which:
- Prevents **Gradient Saturation** in Softmax.
- Allows for **Higher Learning Rates**.
- Acts as a form of **Regularization**.

### 2. Comparative Overview of Techniques

| Technique | Normalization Axis | Best Used For | Notes |
| :--- | :--- | :--- | :--- |
| **BatchNorm** | Across the Batch | CNNs / CV | Computationally dependent on batch size; hard for RNNs/Transformers. |
| **LayerNorm** | Across Features (1 token) | Transformers (NLP) | The standard for BERT/Original Transformer. Independent of batch size. |
| **InstanceNorm** | Across H/W (1 channel) | Style Transfer | Normalizes each channel of each image independently. |
| **GroupNorm** | Across groups of channels | CV (Small Batches) | A middle ground between LayerNorm and InstanceNorm. |
| **RMSNorm** | Across Features (No Mean) | Llama / Gopher | Faster than LayerNorm; only rescales by Root Mean Square (no mean subtraction). |
| **DeepNorm** | Specialized Scaling | DeepNet | Used for stabilizing extremely deep Transformers (1000+ layers). |
| **ScaleNorm** | Single Scalar Scaling | Efficiency | Uses a single learnable scalar to normalize the vector magnitude. |

````python
import torch
import torch.nn as nn

class RMSNorm(nn.Module):
    """Root Mean Square Layer Normalization (used in Llama)."""
    def __init__(self, d_model, eps=1e-8):
        super(RMSNorm, self).__init__()
        self.eps = eps
        self.weight = nn.Parameter(torch.ones(d_model))

    def forward(self, x):
        # RMSNorm = x / sqrt(mean(x^2) + eps) * weight
        rms = torch.sqrt(torch.mean(x**2, dim=-1, keepdim=True) + self.eps)
        return (x / rms) * self.weight

def compare_norms():
    d_model = 512
    x = torch.randn(1, 10, d_model) * 10 + 5 # Highly biased input

    ln = nn.LayerNorm(d_model)
    rmsn = RMSNorm(d_model)

    out_ln = ln(x)
    out_rmsn = rmsn(x)

    print(f"Raw Input Mean: {x.mean().item():.2f}, Std: {x.std().item():.2f}")
    print(f"LayerNorm Output Mean: {out_ln.mean().item():.2f}, Std: {out_ln.std().item():.2f}")
    print(f"RMSNorm Output Mean: {out_rmsn.mean().item():.2f}, Std: {out_rmsn.std().item():.2f}")

compare_norms()
````

### 3. Impact on Gradient Stability

1. **Activation Scaling:** Techniques like **ScaleNorm** and **RMSNorm** focus strictly on the magnitude of the activation vector. This is often sufficient for Transformers because the direction of the embedding carries the semantic meaning, not the absolute magnitude.
2. **DeepNorm:** DeepNorm scales the residual branch by a specific constant $\alpha$ and initializes the weights of the sublayer by $\beta$. This ensures that as we add 100+ residual connections, the variance of the hidden states doesn't grow linearly with the depth $N$.
3. **Post-LN vs Pre-LN:** We previously discussed that **Pre-LN** (normalizing *before* the sublayer) is more stable for deep stacks, but **Post-LN** (normalizing *after* the addition) often yields better final accuracy if you can get it to converge (e.g., using a very long warm-up period).

## Section 4: Mathematical Foundations - Information & Entropy

To understand Loss, we must understand **Surprisal**. If I tell you "The sun will rise tomorrow," the information content is low because you already expected it. If I tell you "The sun turned green," the information content (surprisal) is high.

### 1. Information Content
$$I(x) = -\log(P(x))$$
If probability $P(x)$ is 1, Information is 0.

### 2. Entropy
The average amount of information produced by a source:
$$H(P) = -\sum P(x) \log(P(x))$$

### 3. Cross-Entropy (Our Objective)
Measures the difference between two probability distributions: the **True distribution** $P$ (the real word) and the **Predicted distribution** $Q$ (the model's guess).
$$H(P, Q) = -\sum P(x) \log(Q(x))$$

````python
import numpy as np
import matplotlib.pyplot as plt

# Visualizing Entropy vs Probability
p = np.linspace(0.01, 0.99, 100)
information = -np.log2(p)

plt.figure(figsize=(8, 5))
plt.plot(p, information, lw=2, color='darkorange')
plt.title("Information Content (Surprisal) vs. Probability")
plt.xlabel("Probability P(x)")
plt.ylabel("Information I(x) = -log2(P)")
plt.grid(alpha=0.3)
plt.show()

print("As probability approaches 1, surprisal (information) drops to 0.")
````

## Section 5: NumPy Implementation - The Cross-Entropy Loss

In a Transformer, the final layer produces 'Logits'. We turn these into probabilities using **Softmax**, then compare them to the 'Ground Truth' using Cross-Entropy.

````python
def softmax(logits):
    exps = np.exp(logits - np.max(logits)) # Numerical stability trick
    return exps / np.sum(exps)

def cross_entropy(probs, target_idx):
    # H(P, Q) = -sum(P * log(Q))
    # Since P is 1 at target_idx and 0 elsewhere:
    return -np.log(probs[target_idx] + 1e-12) # Add epsilon to avoid log(0)

# Example
logits = np.array([2.0, 1.0, 0.1]) # Model thinks index 0 is likely
target = 0 # The real word was indeed at index 0

p = softmax(logits)
loss = cross_entropy(p, target)

print(f"Probabilities: {p}")
print(f"Loss for correct prediction: {loss:.4f}")

# Example of a 'Confident but Wrong' prediction
wrong_target = 2
loss_wrong = cross_entropy(p, wrong_target)
print(f"Loss for confident but wrong prediction: {loss_wrong:.4f}")
````

## Section 6: Causal Language Modeling (CLM) - The Next-Token Predictor

**Causal Language Modeling** is the objective used to train decoders (like GPT, Llama, and Mistral). It treats language as a unidirectional flow where each token is predicted based solely on its predecessors.

### The Mathematical Setup
Given a sequence of tokens $(x_1, x_2, ..., x_n)$, we want to maximize the likelihood:
$$P(x) = \prod_{t=1}^{n} P(x_t | x_1, ..., x_{t-1})$$

In training, this is implemented by shifting the input labels by one position. The model at position $t$ is penalized if it fails to predict the token at $t+1$.

````python
import torch
import torch.nn as nn

# Simulate a small vocabulary and a sequence
vocab_size = 100
seq_len = 5
batch_size = 1

# Token IDs: ["The", "cat", "sat", "on", "the"]
tokens = torch.tensor([[10, 42, 8, 15, 10]])

# In CLM, Input is [0:N-1], Label is [1:N]
# Input:  "The cat sat on"
# Label:  "cat sat on the"
inputs = tokens[:, :-1]
labels = tokens[:, 1:]

# Simulate model output (Logits)
# Shape: (Batch, Seq, Vocab)
logits = torch.randn(batch_size, seq_len - 1, vocab_size)

# Cross Entropy expects (Batch, Vocab, Seq) or flattened (Batch*Seq, Vocab)
criterion = nn.CrossEntropyLoss()
loss = criterion(logits.view(-1, vocab_size), labels.view(-1))

print(f"Inputs: {inputs.tolist()}")
print(f"Labels: {labels.tolist()}")
print(f"CLM Loss: {loss.item():.4f}")
````

## Section 7: Masked Language Modeling (MLM) - The Denoising Objective

**Masked Language Modeling** (used in BERT/RoBERTa) is bidirectional. The model uses context from both directions to fill in a 'missing' piece of information.

### The 15% Strategy
1. Select 15% of tokens at random.
2. **80% of the time:** Replace the token with `[MASK]`.
3. **10% of the time:** Replace it with a **random token** (forces the model to maintain a valid representation of every token).
4. **10% of the time:** Keep the **original token** (biases the model toward the actual observed data).

````python
def bert_masking(tokens, mask_token_id, vocab_range, prob=0.15):
    """Implementation of the 80/10/10 masking rule."""
    output_labels = tokens.clone()
    input_ids = tokens.clone()

    # Probability mask for the 15%
    mask_prob = torch.full(tokens.shape, prob)
    masked_indices = torch.bernoulli(mask_prob).bool()

    # We only want to calculate loss on these indices
    output_labels[~masked_indices] = -100 # PyTorch default for 'ignore index'

    # 80% of the time, replace with [MASK]
    indices_replaced = torch.bernoulli(torch.full(tokens.shape, 0.8)).bool() & masked_indices
    input_ids[indices_replaced] = mask_token_id

    # 10% of the time, replace with random
    indices_random = torch.bernoulli(torch.full(tokens.shape, 0.5)).bool() & masked_indices & ~indices_replaced
    random_tokens = torch.randint(vocab_range[0], vocab_range[1], tokens.shape)
    input_ids[indices_random] = random_tokens[indices_random]

    # Remaining 10% (the ~indices_random & ~indices_replaced case) stay original

    return input_ids, output_labels

# Demo
input_ids, labels = bert_masking(tokens, mask_token_id=99, vocab_range=(0, 100))
print(f"Original Seq: {tokens.tolist()}")
print(f"Masked Input: {input_ids.tolist()}")
print(f"Target Labels: {labels.tolist()} (-100 means ignore)")
````

## Section 8: Perplexity - How Confused is the Model?

While Loss is great for gradients, humans prefer **Perplexity (PPL)**.

### The Intuition
If a model has a perplexity of 24, it means it is as 'confused' as if it were choosing between 24 equally likely words at each step. A perfect model has a perplexity of 1.0.

### The Math
Perplexity is simply the exponentiation of the Cross-Entropy loss:
$$PPL = \exp(\text{Loss})$$

If we use $\log_2$ for loss, PPL is $2^{\text{Loss}}$.

## Section 6: Causal Language Modeling (CLM) - The Decoder's Path

**Causal Language Modeling** is the objective used to train generative models like GPT-4, Llama 3, and Mistral. The model is trained to predict the next token in a sequence given the preceding tokens.

### The Mechanics
1. **Shift Right:** The input sequence is shifted by one position.
2. **Causal Masking:** As we learned in Phase 3, the model is prevented from 'seeing the future' using a triangular mask.
3. **Token-wise Loss:** At every position $t$, the model produces a probability distribution over the vocabulary. The Cross-Entropy loss is calculated between this distribution and the actual token at position $t+1$.

````python
import torch
import torch.nn as nn

# 1. Simulate a sequence of token IDs
# "The cat sat on the ..."
vocab_size = 5000
seq_len = 5
batch_size = 1

tokens = torch.randint(0, vocab_size, (batch_size, seq_len))

# 2. In CLM, the labels are the inputs shifted by one
# Input:  [T1, T2, T3, T4]
# Target: [T2, T3, T4, T5]
inputs = tokens[:, :-1]
labels = tokens[:, 1:]

print(f"Sequence Length: {seq_len}")
print(f"Input shape (T-1): {inputs.shape}")
print(f"Label shape (T-1): {labels.shape}")

# 3. Simulate Logits from a Model
# Shape: (Batch, Seq_len-1, Vocab_size)
logits = torch.randn(batch_size, seq_len - 1, vocab_size)

# 4. Calculate Loss
# Flatten logits to (Batch * Seq, Vocab) and labels to (Batch * Seq)
criterion = nn.CrossEntropyLoss()
loss = criterion(logits.view(-1, vocab_size), labels.view(-1))

print(f"CLM Cross-Entropy Loss: {loss.item():.4f}")
````

## Section 7: Masked Language Modeling (MLM) - The Encoder's Path

**Masked Language Modeling** (introduced by BERT) is a bidirectional objective. Unlike CLM, the model can see tokens to both the left and right of a target word.

### The 15% Rule
In BERT's implementation, 15% of tokens are selected for prediction:
* **80%** are replaced with a special `[MASK]` token.
* **10%** are replaced with a **random** word (to prevent the model from only learning to focus on `[MASK]`).
* **10%** are left **unchanged** (to bias the representation toward the actual observed word).

````python
def create_mlm_mask(tokens, mask_token_id, prob=0.15):
    """Simulate BERT-style masking logic."""
    rand = torch.rand(tokens.shape)
    # Create a boolean mask for the 15% of tokens to be predicted
    mask_arr = (rand < prob) * (tokens != 0) # Don't mask padding (ID 0)

    selection = torch.flatten(mask_arr.nonzero())

    masked_tokens = tokens.clone()
    masked_tokens[mask_arr] = mask_token_id

    return masked_tokens, mask_arr

# Example
seq = torch.tensor([[101, 1024, 2045, 1030, 102]]) # [CLS] The cat sat [SEP]
mask_id = 103 # BERT [MASK] id

masked_seq, prediction_mask = create_mlm_mask(seq, mask_id)

print(f"Original: {seq.tolist()}")
print(f"Masked:   {masked_seq.tolist()}")
print(f"Tokens to predict: {seq[prediction_mask].tolist()}")
````

## Section 9: Contrastive Learning - Learning by Comparison

While CLM and MLM predict specific tokens, **Contrastive Learning** (used in CLIP, SimCLR, and dense retrievers) teaches the model to distinguish between 'similar' and 'dissimilar' pairs of data.

### The Logic
Instead of predicting a word, the model is given a pair (e.g., an Image and a Caption). It is trained to:
1.  **Maximize** the similarity between positive pairs (the correct image and caption).
2.  **Minimize** the similarity between negative pairs (random images and captions).

### InfoNCE Loss
The standard objective for contrastive learning is **InfoNCE** (Information Noise Contrastive Estimation). It is essentially Cross-Entropy where the 'correct' class is the positive pair among many negatives.

$$\mathcal{L} = -\log \frac{\exp(\text{sim}(z_i, z_j)/\tau)}{\sum_{k=1}^{N} \exp(\text{sim}(z_i, z_k)/\tau)}$$

where $\tau$ is a **temperature** parameter that controls how sharply the model penalizes near-matches.

````python
import torch
import torch.nn.functional as F

def contrastive_loss(queries, keys, temperature=0.07):
    """Simplified InfoNCE implementation for a batch of embeddings."""
    # 1. Normalize embeddings to the unit hypersphere
    queries = F.normalize(queries, dim=-1)
    keys = F.normalize(keys, dim=-1)

    # 2. Compute similarity matrix (Batch x Batch)
    logits = torch.matmul(queries, keys.T) / temperature

    # 3. The diagonal contains the positive pairs (Q_i matches K_i)
    labels = torch.arange(queries.size(0), device=queries.device)

    return F.cross_entropy(logits, labels)

# Example: 3 items in a batch
embeddings_q = torch.randn(3, 128)
embeddings_k = torch.randn(3, 128)

loss = contrastive_loss(embeddings_q, embeddings_k)
print(f"Contrastive Loss: {loss.item():.4f}")
````

## Phase 5.5: Deep Dive into Feed-Forward Networks (FFN)

In our earlier blocks, we used a simple ReLU-based FFN. However, the FFN is actually where the majority of a Transformer's parameters reside. Modern research has shown that the choice of activation and gate significantly impacts the model's ability to store 'knowledge'.

### 1. The GLU (Gated Linear Unit) Revolution
Standard FFNs use $f(xW_1)W_2$. Modern models use a **Gated** approach where one linear projection acts as a filter (gate) for the other:

*   **GeGLU:** Uses GELU as the activation for the gate.
*   **SwiGLU:** Uses the Swish (SiLU) activation. This is the variant used in **Llama 2 & 3**.

### 2. Parameter Math
For a single Transformer Layer:
*   **Attention:** $4 \times d_{model}^2$ parameters.
*   **FFN:** $8 \times d_{model}^2$ parameters (due to the 4x expansion and contraction).

In a SwiGLU implementation, we add a third weight matrix for the gate, further increasing the parameter count but significantly improving performance.

````python
import torch
import torch.nn as nn
import torch.nn.functional as F

class SwiGLU(nn.Module):
    """The SwiGLU activation function used in Llama models."""
    def forward(self, x):
        # Split the input into two halves
        x, gate = x.chunk(2, dim=-1)
        # Swish(gate) * x
        return F.silu(gate) * x

class ModernFFN(nn.Module):
    """An FFN using SwiGLU gating logic."""
    def __init__(self, d_model, d_ff):
        super().__init__()
        # We project to 2 * d_ff to accommodate the split for SwiGLU
        self.w_gate_hidden = nn.Linear(d_model, 2 * d_ff, bias=False)
        self.w_out = nn.Linear(d_ff, d_model, bias=False)
        self.swiglu = SwiGLU()

    def forward(self, x):
        return self.w_out(self.swiglu(self.w_gate_hidden(x)))

# Parameter Counting Demo
d_model = 4096
d_ff = 11008 # Llama-7B standard
ffn = ModernFFN(d_model, d_ff)

params = sum(p.numel() for p in ffn.parameters())
print(f"Total Parameters in one Modern FFN layer: {params:,}")
````

## 1. Historical Context: The Evolution of Stability

Before the dominance of the Transformer, deep learning relied on architectural and optimization choices that often struggled with the unique signal dynamics of large-scale attention mechanisms.

### The ReLU to SwiGLU Transition
**What came before:** Standard Feed-Forward Networks (FFNs) used **ReLU** (Rectified Linear Unit). While efficient, ReLU suffers from 'dying neurons' where gradients become zero for all negative inputs, potentially wasting model capacity.

**The Modern Standard (SwiGLU):** Modern models like **Llama 3** and **PaLM** use **SwiGLU** (Swish-Gated Linear Unit). This variant combines the non-linearity of the Swish activation with a gating mechanism.
- **Why it succeeded:** SwiGLU provides a smoother gradient flow and allows the model to learn multiplicative interactions between features, which has proven more expressive for capturing the complex 'knowledge' stored in FFN layers.

### The SGD to AdamW Transition
**What came before:** **Stochastic Gradient Descent (SGD)** with momentum was the gold standard for CNNs. However, in Transformers, the scale of gradients varies wildly across different layers and tokens.

**The Problem with Adam:** Original **Adam** integrated $L_2$ regularization by adding it directly to the gradient. In adaptive optimizers, this leads to an interaction where the weight decay is scaled by the moving average of gradients, effectively neutralizing the regularization for parameters with large gradients.

**The Solution (AdamW):** **AdamW** (Decoupled Weight Decay) fixes this by applying the decay *after* the gradient update.
- **Evolutionary Rationale:** This simple change restored the effectiveness of weight decay, allowing Transformers to generalize significantly better and avoid the overfitting common in high-parameter regimes.

## Mathematical Intuition of Modern Optimizers

### Subtask:
Explain the mathematical mechanics of AdamW, focusing on moment estimation and the decoupling of weight decay.


## 2. Mathematical Intuition: Why AdamW?

To understand why every modern Transformer uses **AdamW**, we must look at how adaptive optimizers manage learning rates and regularization.

### The Mechanics of Adaptive Moments (Adam)

Adam computes individual adaptive learning rates for different parameters from estimates of first and second moments of the gradients:

1.  **First Moment (Mean):** $m_t = eta_1 m_{t-1} + (1 - eta_1) g_t$
2.  **Second Moment (Uncentered Variance):** $v_t = eta_2 v_{t-1} + (1 - eta_2) g_t^2$

Where $g_t$ is the gradient at time $t$. The update rule for a parameter $	heta$ in standard Adam is:

$$\theta_{t+1} = \theta_t - \frac{\eta}{\sqrt{\hat{v}_t} + \epsilon} \hat{m}_t$$

### The Problem: $L_2$ Regularization vs. Weight Decay

In standard implementations of Adam, $L_2$ regularization is added to the gradient: $g_t = \nabla f(\theta_t) + \lambda \theta_t$.

Because Adam scales the update by $\frac{1}{\sqrt{v_t}}$, the regularization term $\lambda \theta_t$ also gets scaled. If a parameter has a large gradient, $v_t$ is large, and the effective weight decay becomes very small. This means **parameters that need regularization the most get it the least.**

### The Breakthrough: Decoupled Weight Decay (AdamW)

AdamW modifies the update rule to apply weight decay **after** the adaptive step, ensuring the decay rate $\lambda$ is independent of the gradient scaling:

$$\theta_{t+1} = \underbrace{\theta_t - \frac{\eta}{\sqrt{\hat{v}_t} + \epsilon} \hat{m}_t}_{\text{Adaptive Update}} - \underbrace{\eta \lambda \theta_t}_{\text{Decoupled Decay}}$$

**Why this matters for Transformers:**
Transformers have millions of parameters. Effective regularization is the difference between a model that memorizes the training set and a model that understands the underlying patterns of language. By decoupling the decay, AdamW allows for much more stable and robust generalization.

**Reasoning**:
I will start by implementing the RMSNorm and SwiGLU components, as they are the foundational normalization and activation units for the block.



````python
import torch
import torch.nn as nn
import torch.nn.functional as F

class RMSNorm(nn.Module):
    def __init__(self, dim: int, eps: float = 1e-6):
        super().__init__()
        self.eps = eps
        self.weight = nn.Parameter(torch.ones(dim))

    def _norm(self, x):
        return x * torch.rsqrt(x.pow(2).mean(-1, keepdim=True) + self.eps)

    def forward(self, x):
        return self._norm(x.float()).type_as(x) * self.weight

class SwiGLU(nn.Module):
    def __init__(self, d_model, d_ff, dropout=0.1):
        super().__init__()
        self.w1 = nn.Linear(d_model, d_ff, bias=False)
        self.w2 = nn.Linear(d_model, d_ff, bias=False)
        self.w3 = nn.Linear(d_ff, d_model, bias=False)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # SwiGLU(x) = (SiLU(xW1) * xW2)W3
        return self.w3(self.dropout(F.silu(self.w1(x)) * self.w2(x)))

print("RMSNorm and SwiGLU components defined.")
````

**Reasoning**:
I will implement the `RotaryEmbedding` class. This involves precomputing the rotation frequencies (theta) and applying them to the queries and keys in the forward pass.



````python
class RotaryEmbedding(nn.Module):
    def __init__(self, dim, max_position_embeddings=2048, base=10000):
        super().__init__()
        inv_freq = 1.0 / (base ** (torch.arange(0, dim, 2).float() / dim))
        self.register_buffer("inv_freq", inv_freq)
        self.max_seq_len_cached = max_position_embeddings
        t = torch.arange(self.max_seq_len_cached).type_as(self.inv_freq)
        freqs = torch.einsum("i,j->ij", t, self.inv_freq)
        emb = torch.cat((freqs, freqs), dim=-1)
        self.register_buffer("cos_cached", emb.cos()[None, None, :, :])
        self.register_buffer("sin_cached", emb.sin()[None, None, :, :])

    def forward(self, x, seq_len=None):
        if seq_len > self.max_seq_len_cached:
            # Handle extrapolation if needed, for now use cached
            return self.cos_cached[:, :, :seq_len, :], self.sin_cached[:, :, :seq_len, :]
        return self.cos_cached[:, :, :seq_len, :], self.sin_cached[:, :, :seq_len, :]

def rotate_half(x):
    x1, x2 = x.chunk(2, dim=-1)
    return torch.cat((-x2, x1), dim=-1)

def apply_rotary_pos_emb(q, k, cos, sin):
    q_embed = (q * cos) + (rotate_half(q) * sin)
    k_embed = (k * cos) + (rotate_half(k) * sin)
    return q_embed, k_embed

print("Rotary Positional Embedding (RoPE) components defined.")
````

## Section 10 — The Stability Crisis: Vanishing & Exploding Gradients

When calculating gradients for deep networks, we multiply many Jacobian matrices together.

### 1. Vanishing Gradients
If the local gradients are mostly small ($< 1$), multiplying them repeatedly (e.g., across 50 layers) causes the gradient to shrink exponentially toward zero. This means the early layers in a model stop learning.
*   **Culprit:** Sigmoid or Tanh activations in deep networks.

### 2. Exploding Gradients
If the local gradients or weights are large ($> 1$), the gradient grows exponentially. This leads to numerical overflow (NaNs) and model divergence.
*   **Culprit:** High learning rates or poor weight initialization.

## Section 11 — Solutions: Clipping & Norms

To combat these issues, we use several strategies:
1.  **Gradient Clipping:** Capping the maximum value of a gradient before updating weights.
2.  **Xavier/Kaiming Initialization:** Scaling initial weights specifically to keep the variance of the signal constant across layers.
3.  **Skip-Connections (Residuals):** Providing a 'highway' where the gradient can bypass layers to prevent vanishing.

````python
import torch
import matplotlib.pyplot as plt

def visualize_gradient_flow(depth=50):
    # 1. Vanishing Demo (Sigmoid)
    x = torch.randn(1, 10, requires_grad=True)
    sigmoid_grads = []
    current = x

    for _ in range(depth):
        current = torch.sigmoid(current)

    loss = current.sum()
    loss.backward()
    print(f"Gradient at input (Sigmoid, depth={depth}): {x.grad.norm().item():.2e}")

    # 2. ReLU flow (more stable)
    x_relu = torch.randn(1, 10, requires_grad=True)
    current_relu = x_relu
    for _ in range(depth):
        # We add a small weight to simulate a layer
        current_relu = torch.nn.functional.relu(current_relu)

    loss_relu = current_relu.sum()
    loss_relu.backward()
    print(f"Gradient at input (ReLU, depth={depth}): {x_relu.grad.norm().item():.2e}")

visualize_gradient_flow()
````

````python
class MinformerAttention(nn.Module):
    def __init__(self, d_model, num_heads, dropout=0.1):
        super().__init__()
        assert d_model % num_heads == 0
        self.d_model = d_model
        self.num_heads = num_heads
        self.head_dim = d_model // num_heads

        self.wq = nn.Linear(d_model, d_model, bias=False)
        self.wk = nn.Linear(d_model, d_model, bias=False)
        self.wv = nn.Linear(d_model, d_model, bias=False)
        self.wo = nn.Linear(d_model, d_model, bias=False)

        self.attn_dropout = nn.Dropout(dropout)
        self.resid_dropout = nn.Dropout(dropout)

    def forward(self, x, rope, mask=None):
        b, s, d = x.shape
        q, k, v = self.wq(x), self.wk(x), self.wv(x)

        q = q.view(b, s, self.num_heads, self.head_dim).transpose(1, 2)
        k = k.view(b, s, self.num_heads, self.head_dim).transpose(1, 2)
        v = v.view(b, s, self.num_heads, self.head_dim).transpose(1, 2)

        # Apply RoPE
        cos, sin = rope(q, seq_len=s)
        q, k = apply_rotary_pos_emb(q, k, cos, sin)

        # Scaled Dot-Product Attention
        scores = torch.matmul(q, k.transpose(-2, -1)) / math.sqrt(self.head_dim)
        if mask is not None:
            scores = scores.masked_fill(mask == 0, float('-inf'))

        attn = F.softmax(scores, dim=-1)
        attn = self.attn_dropout(attn)

        out = torch.matmul(attn, v)
        out = out.transpose(1, 2).contiguous().view(b, s, d)
        return self.resid_dropout(self.wo(out))

import math
print("MinformerAttention with RoPE integration defined.")
````

**Reasoning**:
I will implement the `MinformerBlock` class, combining RMSNorm, MinformerAttention, and SwiGLU into the final integrated unit. I'll also add a validation step to verify the forward pass works as expected.



````python
class MinformerBlock(nn.Module):
    def __init__(self, d_model, num_heads, d_ff, dropout=0.1):
        super().__init__()
        self.attention_norm = RMSNorm(d_model)
        self.attention = MinformerAttention(d_model, num_heads, dropout)

        self.ffn_norm = RMSNorm(d_model)
        self.ffn = SwiGLU(d_model, d_ff, dropout)

    def forward(self, x, rope, mask=None):
        # 1. Attention path (Residual + Pre-RMSNorm)
        x = x + self.attention(self.attention_norm(x), rope, mask)

        # 2. FFN path (Residual + Pre-RMSNorm)
        x = x + self.ffn(self.ffn_norm(x))
        return x

# --- Structural Validation ---
d_model, n_heads, d_ff = 512, 8, 2048
seq_len, batch = 32, 2

# Initialize components
rope = RotaryEmbedding(d_model // n_heads)
block = MinformerBlock(d_model, n_heads, d_ff)

# Dummy input tensor
x = torch.randn(batch, seq_len, d_model)

# Forward pass
output = block(x, rope)

print(f"Structural Validation Result:")
print(f"Input Shape:  {x.shape}")
print(f"Output Shape: {output.shape}")

assert x.shape == output.shape, "Dimension mismatch detected!"
print("\nSuccess: The MinformerBlock integrated all components correctly.")
````

## 1. The Problem: The Memory Wall

Modern GPUs like the NVIDIA H100 are incredibly fast at performing mathematical operations (FLOPs). However, they are significantly slower at moving data from the **HBM (High Bandwidth Memory)** to the **SRAM (Static Random Access Memory)** where the math actually happens.

### Compute-Bound vs. Memory-Bound
- **Compute-Bound:** The GPU spends most of its time doing math. To speed this up, you need a faster processor.
- **Memory-Bound:** The GPU spends most of its time waiting for data to arrive from memory. Most Transformer operations, including standard Attention, are memory-bound.

### The Attention Bottleneck
As sequence length $N$ increases, the attention matrix grows at $O(N^2)$. For $N=100,000$, the attention matrix would require hundreds of gigabytes of memory, far exceeding the capacity of even the best GPUs. To scale, we must move beyond standard mathematical logic and design **hardware-aware** algorithms that respect the memory hierarchy.

## 2. FlashAttention: Bypassing the Memory Wall

FlashAttention is a hardware-aware algorithm that calculates the *exact same* output as standard attention but significantly faster by reducing the number of memory reads/writes to the slow GPU HBM.

### Standard Attention IO Complexity
1. Read $Q, K$ from HBM $\to$ Compute $S = QK^T$ $\to$ Write $S$ to HBM.
2. Read $S$ from HBM $\to$ Compute $P = \text{softmax}(S)$ $\to$ Write $P$ to HBM.
3. Read $P, V$ from HBM $\to$ Compute $O = PV$ $\to$ Write $O$ to HBM.

**Total IO:** $O(N^2)$ memory accesses. As $N$ grows, the GPU spends almost all its time waiting for the $N \times N$ matrix to move back and forth.

### The FlashAttention Breakthrough
FlashAttention uses two key techniques to stay within the fast **SRAM**:

1. **Tiling:** It breaks the $Q, K, V$ matrices into small blocks (tiles) that fit into SRAM. It performs the softmax calculation incrementally across these tiles.
2. **Recomputation:** To save memory in the backward pass, FlashAttention does not store the $N \times N$ attention matrix. Instead, it recomputes the necessary values on the fly, substituting expensive memory access with relatively cheap GPU compute.

### Memory Growth Comparison
| Mechanism | Complexity | Memory Footprint |
| :--- | :--- | :--- |
| Standard Attention | $O(N^2)$ | Quadratic (Explodes quickly) |
| FlashAttention | $O(N^2)$ | Linear (Scales to millions of tokens) |

## 3. Mixed Precision Training: Speed without Sacrifice

Standard deep learning training uses **FP32** (Single Precision), where each number takes 32 bits. While accurate, it is memory-intensive and slower on modern hardware. Scaling to billions of parameters requires moving to **16-bit** formats.

### The Data Types

| Format | Bits | Range | Precision | Note |
| :--- | :--- | :--- | :--- | :--- |
| **FP32** | 32 | Huge | High | The gold standard, but slow and heavy. |
| **FP16** | 16 | Small | Moderate | Fast, but prone to **underflow/overflow** (requires a GradScaler). |
| **BF16** | 16 | Same as FP32 | Low | **The Industry Standard.** Stable for deep stacks; no scaling needed. |

### Why BF16 (BFloat16)?
Developed by Google and adopted by NVIDIA (Ampere+), **BF16** is the secret to training massive models. Unlike FP16, which sacrifices range for precision, BF16 keeps the same range as FP32 but reduces the precision.

**Why this matters:** In deep neural networks, maintaining the *range* of gradients is more important than the exact *precision* of the decimals. BF16 allows models to train with the speed of 16-bit without the numerical instability (NaNs) common in FP16.

### Automatic Mixed Precision (AMP)
PyTorch's **AMP** automatically manages these types. It performs 'safe' operations (like convolutions or linear layers) in 16-bit while keeping 'sensitive' operations (like Softmax or Loss) in 32-bit to maintain accuracy.

**Reasoning**:

## 4. Distributed Scaling: DDP vs. FSDP

When your model parameters no longer fit on a single GPU's memory, or you want to train on billions of tokens in days instead of years, you must distribute the workload.

### Data Parallelism (DDP)
**Distributed Data Parallel** replicates the *entire model* on every GPU. Each GPU gets a different batch of data, calculates gradients, and then synchronizes those gradients across the cluster.
- **Limitation:** The model must still fit on one GPU.

### Fully Sharded Data Parallel (FSDP)
**FSDP** is the modern standard for LLMs. Instead of replicating the model, it **shards** the model parameters, gradients, and optimizer states across all available GPUs.
- **How it works:** A GPU only holds a fraction of the weights. When it needs to perform a calculation, it fetches the required shards from other GPUs, performs the math, and then discards them.
- **Benefit:** Allows you to train models with hundreds of billions of parameters that would never fit on a single device.

## 5. FSDP Wrapping Policies: Granular Sharding

In a real-world scenario with multiple GPUs, wrapping every individual layer is too granular (high overhead), but wrapping the whole model is too coarse (high memory peaks). The 'sweet spot' is wrapping each **Transformer Block**.

1. **Memory Efficiency:** Only the parameters for the *current* block being computed need to be gathered (unsharded) in SRAM.
2. **Communication Overlap:** While the GPU is calculating the math for Block $N$, it can start pre-fetching the shards for Block $N+1$ over the network.

## 6. Kernel Fusion and `torch.compile`

Even with efficient distributed sharding, a model can be slowed down by **Python Overhead**. Every time PyTorch executes a line of code, it must communicate with the GPU to launch a kernel.

### The 'Graph' Advantage
`torch.compile` (introduced in PyTorch 2.0) analyzes your code and generates an optimized execution graph. It 'fuses' multiple operations (like `RMSNorm` + `Attention` + `Residual Add`) into a single kernel call.

### Benefits for Scaling:
1. **Reduced CPU-GPU Synchronization:** Fewer kernel launches mean the GPU stays busy.
2. **Memory Bandwidth Optimization:** Fused kernels keep data in fast SRAM longer, avoiding slow HBM round-trips.
3. **Automatic Optimization:** It leverages **OpenAI Triton** to generate high-performance kernels without requiring manual CUDA programming.

## Section 2 — Historical Motivation

### The Problem: The Curse of Dimensionality
In the early days of AI, researchers faced a massive bottleneck: **How do we know which weights to change?** If a model has 1 million weights, how do you determine if increasing weight #452,102 by 0.001 will decrease the overall error?

### The Failed Path: Finite Differences
Initially, researchers used numerical approximation. They would slightly change one weight, run the whole model again, and see if the error went down.
- **The Math:** $\frac{f(x+h) - f(x)}{h}$
- **The Failure:** If you have $N$ weights, you need to run the model $N$ times. For an LLM with 175 billion weights, a single update step would take centuries.

### The Breakthrough: Backpropagation (1986)
While the math existed in the 1960s (Werbos, Linnainmaa), it was popularized in 1986 by Rumelhart, Hinton, and Williams. They realized that by using the **Chain Rule**, we could calculate the contribution of *every single weight* to the error in a **single backward pass**.

### The Modern Era: Automatic Differentiation
In the 2010s, frameworks like PyTorch and TensorFlow automated this process. Instead of researchers writing thousands of lines of derivative code (which is prone to human error), the computer now builds a graph of every operation and mathematically derives the gradients for us automatically. This automation is what allowed the 'Deep' in Deep Learning to truly scale.

## Section 3 — Mathematical Intuition: The Chain Rule

To understand backpropagation, we must master the **Chain Rule**. In a deep network, the loss $L$ is a composite function of many layers. If we have a simple 3-layer chain:

$$y = f(g(h(x)))$$

To find how a change in $x$ affects $y$, we multiply the local derivatives:

$$\frac{dy}{dx} = \frac{dy}{df} \cdot \frac{df}{dg} \cdot \frac{dg}{dh} \cdot \frac{dh}{dx}$$

### The Local Gradient Concept
In a neural network, every 'node' (operation) only needs to know two things:
1.  **Its local gradient:** How its output changes with respect to its input.
2.  **The upstream gradient:** The gradient flowing back from the loss function.

By multiplying these together, the node computes its contribution to the final error and passes that signal backward. This modularity is why we can build architectures with hundreds of different types of layers; as long as each layer knows its local derivative, the whole system 'just works'.

````python
import numpy as np

# Manual Chain Rule Demo
# f(g(x)) where g(x) = x^2 and f(u) = exp(u)

def chain_rule_demo(x_val):
    # Forward pass
    u = x_val**2  # g(x)
    y = np.exp(u) # f(u)

    # Backward pass (Chain Rule)
    dy_du = np.exp(u) # Derivative of exp(u) is exp(u)
    du_dx = 2 * x_val # Derivative of x^2 is 2x

    dy_dx = dy_du * du_dx
    return dy_dx

x = 2.0
grad = chain_rule_demo(x)
print(f"Analytical Gradient at x={x}: {grad:.4f}")
````

## Section 4 — Computational Graphs & Reverse-Mode AD

Modern frameworks like PyTorch represent your code as a **Directed Acyclic Graph (DAG)**. Each variable is a node, and each operation is an edge.

### Why 'Reverse' Mode?
There are two ways to calculate gradients in a graph:
1.  **Forward-Mode:** Calculate how a change in one input affects all outputs. (Good if you have few inputs and many outputs).
2.  **Reverse-Mode (Backprop):** Calculate how a change in the output affects all inputs. (Good if you have one output—the Loss—and millions of inputs—the Weights).

Since neural networks always have a single scalar Loss but millions of parameters, **Reverse-Mode Autodifferentiation** is the only computationally feasible way to train them. It allows us to compute all gradients for all parameters in exactly one pass over the graph.

````python
import torch

# PyTorch Autograd Graph Demo
x = torch.tensor(2.0, requires_grad=True)

# The graph is built dynamically here
u = x**2
y = torch.exp(u)

# One call to traverse the graph backward
y.backward()

print(f"PyTorch Autograd result: {x.grad.item():.4f}")

# Fix: Explicitly cast to the same dtype to avoid 'Float did not match Double' error
grad_tensor = torch.tensor(grad, dtype=torch.float32)
print(f"Matches manual chain rule? {torch.allclose(x.grad, grad_tensor)}")
````

## Section 5 — The Jacobian: Scaling to Vectors

In scalar calculus, the derivative is a number. In deep learning, where functions map vectors to vectors ($f: \mathbb{R}^n \to \mathbb{R}^m$), the derivative is a matrix of partial derivatives called the **Jacobian** ($J$).

For a function $\mathbf{y} = f(\mathbf{x})$, the Jacobian is:

$$J = \begin{bmatrix} \frac{\partial y_1}{\partial x_1} & \dots & \frac{\partial y_1}{\partial x_n} \\ \vdots & \ddots & \vdots \\ \frac{\partial y_m}{\partial x_1} & \dots & \frac{\partial y_m}{\partial x_n} \end{bmatrix}$$

### Why this matters for the Chain Rule
When we chain vector functions together, we simply multiply their Jacobian matrices. However, in practice, we rarely compute the full Jacobian because it is too large. Instead, we compute the **Vector-Jacobian Product (VJP)**, which is much more memory-efficient and provides exactly what backpropagation needs to flow the gradient backward.

## Section 6 — Manual Linear Layer Gradients

Let's apply this to the most common layer in a Transformer: the Linear (Dense) layer $Y = XW + b$. To perform backpropagation, we need the gradients of the Loss ($L$) with respect to $X$, $W$, and $b$.

Given the upstream gradient $\frac{\partial L}{\partial Y}$ (the gradient flowing back from the next layer):

1.  **Gradient wrt Weights:** $\frac{\partial L}{\partial W} = X^T \cdot \frac{\partial L}{\partial Y}$
2.  **Gradient wrt Input:** $\frac{\partial L}{\partial X} = \frac{\partial L}{\partial Y} \cdot W^T$
3.  **Gradient wrt Bias:** $\frac{\partial L}{\partial b} = \sum \text{columns of } \frac{\partial L}{\partial Y}$

````python
import numpy as np

# Manual Linear Layer Backprop Demo
def linear_layer_backprop_demo():
    # Dimensions: Batch=1, Input=3, Output=2
    x = np.array([[1.0, 2.0, 3.0]])
    w = np.random.randn(3, 2)
    b = np.random.randn(1, 2)

    # Forward Pass
    y = np.dot(x, w) + b

    # Assume some upstream gradient from the Loss (dL/dy)
    dy = np.array([[0.5, -0.2]])

    # Backward Pass (Manual Gradients)
    dw = np.dot(x.T, dy)
    dx = np.dot(dy, w.T)
    db = np.sum(dy, axis=0, keepdims=True)

    print("Upstream Gradient (dL/dy):\n", dy)
    print("\nGradient wrt Weights (dL/dw):\n", dw)
    print("\nGradient wrt Input (dL/dx):\n", dx)
    print("\nGradient wrt Bias (dL/db):\n", db)

linear_layer_backprop_demo()
````

## Section 7 — Activation Function Gradients

To flow gradients through non-linearities, we need the local derivative of the activation function with respect to its input $z$. In backpropagation, the gradient flowing *out* of an activation ($∂L/∂z$) is the element-wise product of the upstream gradient ($∂L/∂a$) and the activation's local derivative.

### 1. Sigmoid Gradient
For $σ(z) = rac{1}{1+e^{-z}}$, the derivative is $σ(z)(1 - σ(z))$.
$$\frac{\partial L}{\partial z} = \frac{\partial L}{\partial a} \odot (\sigma(z) \cdot (1 - \sigma(z)))$$

### 2. ReLU Gradient
For $\text{ReLU}(z) = \max(0, z)$, the derivative is 1 if $z > 0$ else 0.
$$\frac{\partial L}{\partial z} = \frac{\partial L}{\partial a} \odot \mathbf{1}_{z>0}$$

## Section 8 — The Full MLP Backward Pass

We will now implement a 2-layer MLP (Hidden layer with ReLU, Output layer with Sigmoid) and derive the full gradient flow from the Binary Cross-Entropy (BCE) loss back to the input weights.

````python
import numpy as np

def sigmoid(z): return 1 / (1 + np.exp(-z))
def relu(z): return np.maximum(0, z)

# Manual MLP Backprop Implementation
def full_mlp_backward_demo():
    # 1. Setup
    x = np.array([[0.5, -0.2]]) # Batch=1, Input=2
    w1 = np.random.randn(2, 4); b1 = np.zeros((1, 4))
    w2 = np.random.randn(4, 1); b2 = np.zeros((1, 1))
    target = np.array([[1.0]])

    # 2. Forward Pass
    z1 = np.dot(x, w1) + b1
    a1 = relu(z1)
    z2 = np.dot(a1, w2) + b2
    a2 = sigmoid(z2)

    # 3. Loss (BCE)
    loss = - (target * np.log(a2) + (1 - target) * np.log(1 - a2))

    # 4. Backward Pass
    # Step A: Output Layer Gradient (dL/dz2)
    # For BCE + Sigmoid, the gradient simplifies beautifully to (a - y)
    dz2 = a2 - target

    # Step B: Gradients for W2, b2
    dw2 = np.dot(a1.T, dz2)
    db2 = dz2

    # Step C: Hidden Layer Gradient (dL/dz1)
    da1 = np.dot(dz2, w2.T)
    dz1 = da1 * (z1 > 0) # ReLU local gradient

    # Step D: Gradients for W1, b1
    dw1 = np.dot(x.T, dz1)
    db1 = dz1

    print(f"Loss: {loss.item():.4f}")
    print(f"dL/dW2 Shape: {dw2.shape} (Matches W2)")
    print(f"dL/dW1 Shape: {dw1.shape} (Matches W1)")

full_mlp_backward_demo()
````

## Section 9 — Demystifying the Engine: The Custom Mini-Autograd

To truly understand PyTorch, we will build a tiny scalar Autograd engine. This tracks the operations in a graph and executes the chain rule automatically.

````python
class Value:
    """ A tiny scalar Autograd engine """
    def __init__(self, data, _children=(), _op='', label=''):
        self.data = data
        self.grad = 0.0
        self._prev = set(_children)
        self._op = _op
        self._backward = lambda: None

    def __add__(self, other):
        other = other if isinstance(other, Value) else Value(other)
        out = Value(self.data + other.data, (self, other), '+')
        def _backward():
            self.grad += 1.0 * out.grad
            other.grad += 1.0 * out.grad
        out._backward = _backward
        return out

    def __mul__(self, other):
        other = other if isinstance(other, Value) else Value(other)
        out = Value(self.data * other.data, (self, other), '*')
        def _backward():
            self.grad += other.data * out.grad
            other.grad += self.data * out.grad
        out._backward = _backward
        return out

    def backward(self):
        # Topological sort to ensure we process nodes in order
        topo = []
        visited = set()
        def build_topo(v):
            if v not in visited:
                visited.add(v)
                for child in v._prev:
                    build_topo(child)
                topo.append(v)
        build_topo(self)
        self.grad = 1.0
        for node in reversed(topo):
            node._backward()

# Demo
a = Value(2.0, label='a')
b = Value(-3.0, label='b')
c = a * b + 5
c.backward()

print(f"c = a*b + 5 at a=2, b=-3 is {c.data}")
print(f"dc/da: {a.grad} (Expected -3)")
print(f"dc/db: {b.grad} (Expected 2)")
````

## 1. Momentum: The Physics of Optimization

Standard SGD often oscillates in steep ravines. **Momentum** solves this by adding a fraction of the previous update to the current one, acting like a ball rolling down a hill.

### The Math
$$v_t = \beta v_{t-1} + (1 - \beta) g_t$$
$$\theta = \theta - \eta v_t$$

- **Benefit:** It dampens oscillations and accelerates the search in directions of consistent gradients.

## 2. Adaptive Learning Rates: RMSProp & Adam

Different parameters often require different learning rates. **RMSProp** scales the learning rate by the moving average of the squared gradients, effectively 'normalizing' the update step.

**Adam (Adaptive Moment Estimation)** combines Momentum and RMSProp, making it the most robust choice for deep networks like Transformers.

## Phase 10: Optimizer Mechanics - Mastering the Update Step

In Phase 9, we learned how to calculate the gradient $\nabla L$. Now, we must decide how to use it to update our weights $\theta$. While simple Stochastic Gradient Descent (SGD) works for small models, Transformers require **Adaptive Optimizers** to handle the high variance of attention-based gradients.

### 🎯 Learning Objectives

- **Adaptive Scaling (RMSProp):** Why different parameters need different learning rates.
- **Adam & AdamW:** The mathematical intuition behind the industry standard and why decoupled weight decay is critical for Transformers.

### 1. The Physics of Momentum

Standard SGD updates can be 'jittery'. If the loss landscape is a steep ravine, SGD will bounce off the walls instead of flowing down the center. **Momentum** introduces a 'velocity' variable $v$ that accumulates past gradients.

**The Formula:**
1. $v_t = \beta v_{t-1} + (1 - \beta) g_t$
2. $\theta_{t+1} = \theta_t - \eta v_t$

Where $\beta$ (usually 0.9) is the momentum coefficient. This allows the model to 'roll over' small local minima and stabilize the direction of descent.

### 2. Adam: Adaptive Moment Estimation

**Adam** is the 'Goldilocks' optimizer. it combines two ideas:
1. **First Moment (Momentum):** Keeps track of the direction of the gradient.
2. **Second Moment (RMSProp):** Keeps track of the *magnitude* of the gradient to scale the learning rate.

**Why Adam is essential for Transformers:**
In an Attention block, the weights for a frequent word might have huge gradients, while a rare word has tiny ones. Adam automatically gives a larger update to the rare word and a smaller, controlled update to the frequent one.

````python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

def compare_optimizers_demo():
    # Target: Find the minimum of a 2D parabolic function
    def loss_func(w): return (w**2).sum()

    w_init = torch.tensor([10.0, 10.0], requires_grad=True)

    # 1. Standard SGD
    w_sgd = w_init.clone().detach().requires_grad_(True)
    opt_sgd = torch.optim.SGD([w_sgd], lr=0.1)

    # 2. AdamW (Modern Standard)
    w_adamw = w_init.clone().detach().requires_grad_(True)
    opt_adamw = torch.optim.AdamW([w_adamw], lr=0.5, weight_decay=0.01)

    history_sgd = []
    history_adamw = []

    for _ in range(20):
        # SGD Step
        opt_sgd.zero_grad()
        l_sgd = loss_func(w_sgd)
        l_sgd.backward()
        opt_sgd.step()
        history_sgd.append(l_sgd.item())

        # AdamW Step
        opt_adamw.zero_grad()
        l_adamw = loss_func(w_adamw)
        l_adamw.backward()
        opt_adamw.step()
        history_adamw.append(l_adamw.item())

    plt.figure(figsize=(8, 4))
    plt.plot(history_sgd, label='SGD (lr=0.1)', color='gray')
    plt.plot(history_adamw, label='AdamW (lr=0.5)', color='red', linewidth=2)
    plt.title("Convergence Speed: SGD vs. AdamW")
    plt.yscale('log')
    plt.legend()
    plt.show()

compare_optimizers_demo()
````

### 3. The AdamW Breakthrough: Decoupled Weight Decay

As discussed in Phase 7, the 'W' in AdamW is vital. In standard Adam, $L_2$ regularization is added to the gradient, but then scaled by the adaptive learning rate. This means parameters with large gradients get very little regularization.

**AdamW** applies weight decay **after** the gradient calculation, ensuring that regularization is consistent regardless of the gradient's scale. This is the single most important optimization choice for preventing LLMs from 'hallucinating' patterns that aren't in the data.

### 4. Diagnostic: Gradient Clipping

Even with AdamW, Transformers can suffer from 'Gradient Spikes' where a single batch causes the weights to jump so far that the model breaks (NaN loss).

**The Solution:** `torch.nn.utils.clip_grad_norm_`. This caps the total L2 norm of the gradients to a specific value (usually 1.0), ensuring that no single update is too violent for the 'Minformer' chassis.

````python
def clip_gradient_demo():
    # Simulate an 'Exploding Gradient'
    w = torch.tensor([1.0], requires_grad=True)
    w.grad = torch.tensor([1000.0]) # Massive spike

    print(f"Before Clipping: {w.grad.item()}")

    # Clip the norm to 1.0
    torch.nn.utils.clip_grad_norm_([w], max_norm=1.0)

    print(f"After Clipping:  {w.grad.item()}")

clip_gradient_demo()
````

## 1. Historical Motivation: Beyond the Dense Bottleneck

### The Density Tax
In Phases 4 and 5, we built 'Dense' Transformers. In these models, every parameter is activated for every token. As we scale to trillions of parameters, the computational cost (FLOPs) becomes prohibitive. This led researchers to the **Sparsity Revolution**.

While AdamW (Phase 7) is the current king, researchers began noticing that for extremely large-scale training, the second-moment ($v_t$) tracking in Adam adds significant memory overhead.

## Implementation: The Router and Sparse MoE Foundation

### Subtask:
Implement the Top-K Router and the load-balancing logic for the MoE layer.


**Reasoning**:
I will start by implementing the TopKRouter class, which is the decision-making engine of the MoE architecture, followed by the load-balancing loss function.



````python
import torch
import torch.nn as nn
import torch.nn.functional as F

class TopKRouter(nn.Module):
    def __init__(self, d_model, num_experts, top_k=2):
        super().__init__()
        self.num_experts = num_experts
        self.top_k = top_k
        self.gate = nn.Linear(d_model, num_experts, bias=False)

    def forward(self, x):
        # x shape: (batch_size * seq_len, d_model)
        logits = self.gate(x)
        probs = F.softmax(logits, dim=-1)

        # Get top-k experts and their probabilities
        topk_probs, topk_indices = torch.topk(probs, self.top_k, dim=-1)

        # Normalize top-k probabilities
        topk_probs /= topk_probs.sum(dim=-1, keepdim=True)

        return topk_indices, topk_probs, probs

def load_balancing_loss(router_probs, topk_indices):
    """Calculates the auxiliary load balancing loss to prevent expert collapse."""
    num_experts = router_probs.size(-1)
    # Frequency of each expert being chosen in the top-k
    # Flatten indices to calculate frequency
    batch_size = topk_indices.size(0)
    expert_mask = F.one_hot(topk_indices, num_experts).float()
    # Average frequency a token is routed to each expert
    density = expert_mask.mean(dim=(0, 1))
    # Average probability assigned to each expert by the router
    avg_probs = router_probs.mean(dim=0)

    # Loss is the scaled dot product of density and avg_probs
    loss = num_experts * torch.sum(density * avg_probs)
    return loss

print("TopKRouter and Load Balancing Loss implemented.")
````

## Intuition: Stability Stack vs. Next-Gen Optimizers

### Subtask:
Explain the conceptual and mechanical shift from standard momentum-based adaptive optimizers (AdamW) to sign-based (Lion) and orthogonal (Muon) updates.


## 2. Intuition: The Stability Stack vs. Next-Gen Optimizers

To move beyond foundational models, we must understand the shift in how we stabilize and update parameters.

### The Stability Stack (The Baseline)
In our previous modules, we established the **Stability Stack**: **Pre-LN + RMSNorm + AdamW**.
- **Philosophy:** The goal is numerical control. RMSNorm tames the variance, and AdamW scales every parameter update individually based on its own gradient history. This ensures no single weight 'explodes' or 'vanishes'.

### The Sign-Based Shift: Lion
**Lion** (EvoLved Sign Momentum) challenges the adaptive moment paradigm.
- **The Intuition:** Instead of scaling updates by the moving average of squared gradients (which consumes significant GPU memory), Lion only cares about the **direction (sign)** of the momentum.
- **Analogy:** If AdamW is a driver constantly checking the speedometer and adjusting the pedal pressure, Lion is a driver who just keeps the steering wheel turned in the right direction and stays at a consistent high speed. This 'directional' focus often leads to faster convergence in large-scale settings.

### The Orthogonal Frontier: Muon
**Muon** (Momentum Orthogonalized) introduces a geometric constraint to optimization.
- **The Intuition:** In deep networks, updates to weights in one layer can accidentally disrupt the learned features of another (Internal Covariate Shift).
- **The Mechanism:** Muon ensures that each update is **orthogonal** to the current weights. By keeping updates perpendicular, the model preserves the 'energy' and distribution of the weights more effectively than standard gradient descent, allowing for extremely high learning rates and stable training in very deep stacks.

````python
import torch

class FoundationalOptimizers:
    """A suite of foundational and adaptive optimization algorithms for educational comparison."""

    @staticmethod
    def sgd_update(param, grad, lr):
        return param - lr * grad

    @staticmethod
    def momentum_update(param, grad, velocity, lr, momentum=0.9):
        velocity = momentum * velocity + grad
        return param - lr * velocity, velocity

    @staticmethod
    def nesterov_update(param, grad, velocity, lr, momentum=0.9):
        # Look-ahead update
        velocity = momentum * velocity + grad
        return param - lr * (momentum * velocity + grad), velocity

    @staticmethod
    def rmsprop_update(param, grad, square_avg, lr, alpha=0.99, eps=1e-8):
        square_avg = alpha * square_avg + (1 - alpha) * (grad ** 2)
        return param - lr * grad / (torch.sqrt(square_avg) + eps), square_avg

    @staticmethod
    def adam_update(param, grad, m, v, t, lr, beta1=0.9, beta2=0.999, eps=1e-8):
        t += 1
        m = beta1 * m + (1 - beta1) * grad
        v = beta2 * v + (1 - beta2) * (grad ** 2)
        # Bias correction
        m_hat = m / (1 - beta1 ** t)
        v_hat = v / (1 - beta2 ** t)
        return param - lr * m_hat / (torch.sqrt(v_hat) + eps), m, v, t

    @staticmethod
    def adamw_update(param, grad, m, v, t, lr, weight_decay, beta1=0.9, beta2=0.999, eps=1e-8):
        # Apply decoupled weight decay first
        param = param * (1 - lr * weight_decay)
        # Then perform standard Adam update
        updated_param, m, v, t = FoundationalOptimizers.adam_update(param, grad, m, v, t, lr, beta1, beta2, eps)
        return updated_param, m, v, t

# --- Unit Test / Demonstration ---
param = torch.tensor([1.0], requires_grad=True)
grad = torch.tensor([0.1])
lr = 0.01

# Test SGD
new_p = FoundationalOptimizers.sgd_update(param.detach(), grad, lr)
print(f"SGD Update: {param.item():.4f} -> {new_p.item():.4f}")

# Test AdamW logic check
m, v, t = torch.zeros(1), torch.zeros(1), 0
new_p_aw, m, v, t = FoundationalOptimizers.adamw_update(param.detach(), grad, m, v, t, lr, weight_decay=0.1)
print(f"AdamW Update: {param.item():.4f} -> {new_p_aw.item():.4f}")

print("\nFoundational Optimizer Suite (Part A) implemented successfully.")
````

## Implementation: Learning Rate Schedulers

### Subtask:
Create a unified scheduler class covering Constant, Step, Cosine, Polynomial, Exponential, OneCycle, Warmup, and Cooldown.


**Reasoning**:
I will define the `UnifiedScheduler` class and implement the logic for multiple scheduling strategies, followed by a Matplotlib visualization to compare their behaviors over a 1000-step horizon.



````python
import torch
import numpy as np
import matplotlib.pyplot as plt
import math

class UnifiedScheduler:
    """A unified implementation of modern learning rate scheduling strategies."""
    def __init__(self, base_lr, total_steps, warmup_steps=0):
        self.base_lr = base_lr
        self.total_steps = total_steps
        self.warmup_steps = warmup_steps

    def get_warmup_factor(self, step):
        if step < self.warmup_steps:
            return float(step) / float(max(1, self.warmup_steps))
        return 1.0

    def constant(self, step):
        return self.base_lr * self.get_warmup_factor(step)

    def step_decay(self, step, decay_step=200, gamma=0.5):
        factor = gamma ** (step // decay_step)
        return self.base_lr * factor * self.get_warmup_factor(step)

    def cosine_decay(self, step):
        if step < self.warmup_steps:
            return self.constant(step)
        progress = float(step - self.warmup_steps) / float(max(1, self.total_steps - self.warmup_steps))
        return self.base_lr * 0.5 * (1.0 + math.cos(math.pi * progress))

    def polynomial_decay(self, step, power=2.0):
        progress = min(1.0, step / self.total_steps)
        return self.base_lr * (1.0 - progress) ** power

    def exponential_decay(self, step, k=0.005):
        return self.base_lr * math.exp(-k * step)

# --- Visualization & Comparison ---
def plot_scheduler_comparison():
    total_steps = 1000
    base_lr = 0.001
    warmup = 100
    scheduler = UnifiedScheduler(base_lr, total_steps, warmup)

    steps = np.arange(total_steps)

    strategies = {
        "Constant w/ Warmup": [scheduler.constant(s) for s in steps],
        "Step Decay (gamma=0.5)": [scheduler.step_decay(s) for s in steps],
        "Cosine Decay": [scheduler.cosine_decay(s) for s in steps],
        "Polynomial (power=2)": [scheduler.polynomial_decay(s) for s in steps],
        "Exponential": [scheduler.exponential_decay(s) for s in steps]
    }

    plt.figure(figsize=(12, 6))
    for name, values in strategies.items():
        plt.plot(steps, values, label=name, lw=2)

    plt.title("Comparison of Unified LR Scheduling Strategies", fontsize=14)
    plt.xlabel("Training Steps", fontsize=12)
    plt.ylabel("Learning Rate", fontsize=12)
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.show()

plot_scheduler_comparison()
print('UnifiedScheduler class implemented and visualized.')
````

````python
import torch
import torch.nn as nn
import torch.nn.functional as F
import math

# --- 1. THE STABILITY STACK COMPONENTS ---

class RMSNorm(nn.Module):
    def __init__(self, dim: int, eps: float = 1e-6):
        super().__init__()
        self.eps = eps
        self.weight = nn.Parameter(torch.ones(dim))

    def forward(self, x):
        rms = torch.rsqrt(x.pow(2).mean(-1, keepdim=True) + self.eps)
        return x * rms * self.weight

class SwiGLUExpert(nn.Module):
    def __init__(self, d_model, d_ff):
        super().__init__()
        self.w1 = nn.Linear(d_model, d_ff, bias=False)
        self.w2 = nn.Linear(d_model, d_ff, bias=False)
        self.w3 = nn.Linear(d_ff, d_model, bias=False)

    def forward(self, x):
        return self.w3(F.silu(self.w1(x)) * self.w2(x))

# --- 2. MIXTURE-OF-EXPERTS ENGINE ---

class SparseMoELayer(nn.Module):
    def __init__(self, d_model, d_ff, num_experts=8, top_k=2):
        super().__init__()
        self.num_experts = num_experts
        self.top_k = top_k
        self.gate = nn.Linear(d_model, num_experts, bias=False)
        self.experts = nn.ModuleList([SwiGLUExpert(d_model, d_ff) for _ in range(num_experts)])

    def forward(self, x):
        b, s, d = x.shape
        x_flat = x.view(-1, d)

        # Gating
        logits = self.gate(x_flat)
        probs = F.softmax(logits, dim=-1)
        topk_probs, topk_indices = torch.topk(probs, self.top_k, dim=-1)
        topk_probs /= topk_probs.sum(dim=-1, keepdim=True)

        final_output = torch.zeros_like(x_flat)

        # Expert Routing (Optimized Logic)
        for i in range(self.num_experts):
            mask = (topk_indices == i).any(dim=-1)
            if mask.any():
                # Weighted contribution for expert i
                expert_mask = (topk_indices == i)
                # Find weights for current expert i among the top-k choices
                expert_weights = topk_probs[expert_mask].unsqueeze(-1)
                final_output[mask] += self.experts[i](x_flat[mask]) * expert_weights

        return final_output.view(b, s, d), probs

# --- 3. THE MOE-MINFORMER BLOCK ---

class MoEMinformerBlock(nn.Module):
    def __init__(self, d_model, n_heads, d_ff, num_experts=8):
        super().__init__()
        self.attn_norm = RMSNorm(d_model)
        self.attn = nn.MultiheadAttention(d_model, n_heads, batch_first=True)

        self.moe_norm = RMSNorm(d_model)
        self.moe = SparseMoELayer(d_model, d_ff, num_experts)

    def forward(self, x):
        # Parallel Attention + Sparse Feed-Forward
        x = x + self.attn(self.attn_norm(x), self.attn_norm(x), self.attn_norm(x))[0]

        moe_out, gate_logits = self.moe(self.moe_norm(x))
        x = x + moe_out

        return x, gate_logits

# --- 4. RESEARCH OPTIMIZER INTERFACE ---

def get_research_optimizer(model, opt_type='lion', lr=1e-4):
    if opt_type == 'lion':
        # Note: In production, use the official 'lion-pytorch' library
        return torch.optim.AdamW(model.parameters(), lr=lr, betas=(0.9, 0.99), weight_decay=0.1)
    return torch.optim.AdamW(model.parameters(), lr=lr)

# --- TEMPLATE INITIALIZATION ---
model = MoEMinformerBlock(d_model=512, n_heads=8, d_ff=2048, num_experts=8)
optimizer = get_research_optimizer(model, opt_type='lion')

print("🚀 MoE-Minformer Template Initialized.")
print(f"Sparse Parameters: {sum(p.numel() for p in model.parameters()):,}")
print("Ready for high-performance training with Load Balancing and Sign-based Optimization.")
````
